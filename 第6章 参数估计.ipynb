### 习题 6 详细解答

#### 1. 两点分布参数 $p$ 的估计
**解题思路**：利用矩估计法（一阶原点矩等于样本均值）和最大似然估计法（对数似然函数求导）。
**解答**：
总体 $X \sim B(1, p)$，即 $P(X=1)=p, P(X=0)=1-p$。
(1) **矩估计**：$E(X) = p$。令 $E(X) = \bar{X}$，解得 $p$ 的矩估计量为 $\hat{p}_{1} = \bar{X}$。
(2) **最大似然估计**：似然函数 $L(p) = \prod_{i=1}^n p^{X_i}(1-p)^{1-X_i} = p^{\sum X_i}(1-p)^{n-\sum X_i}$。
取对数 $\ln L(p) = (\sum X_i)\ln p + (n-\sum X_i)\ln(1-p)$。
求导并令其为0：$\frac{\sum X_i}{p} - \frac{n-\sum X_i}{1-p} = 0$，解得 $\hat{p}_{2} = \frac{1}{n}\sum_{i=1}^n X_i = \bar{X}$。

#### 2. 概率密度函数参数 $\theta$ 的估计
**解题思路**：计算总体一阶原点矩，令其等于样本均值。写出似然函数求导。
**解答**：
总体的概率密度 $f(x, \theta) = (\theta+1)x^\theta, 0<x<1$。
(1) **矩估计**：$E(X) = \int_0^1 x(\theta+1)x^\theta dx = \frac{\theta+1}{\theta+2}$。令 $\frac{\theta+1}{\theta+2} = \bar{X}$，解得 $\hat{\theta}_{1} = \frac{2\bar{X}-1}{1-\bar{X}}$。
(2) **最大似然估计**：似然函数 $L(\theta) = (\theta+1)^n \prod_{i=1}^n x_i^\theta$。
对数似然 $\ln L = n\ln(\theta+1) + \theta\sum_{i=1}^n \ln x_i$。求导令为0：$\frac{n}{\theta+1} + \sum\ln x_i = 0$，解得 $\hat{\theta}_{2} = -\frac{n}{\sum_{i=1}^n \ln x_i} - 1$。

#### 3. 最大似然估计量的选择题
**解题思路**：利用极大似然估计的不变性原理。
**解答**：
总体 $X \sim N(\mu, \sigma^2)$，$\mu$ 的极大似然估计量为 $\bar{X}$。由极大似然估计的不变性原理，$2\mu+1$ 的极大似然估计量为 $2\bar{X}+1$。
**答案**：(A)

#### 4. 离散型分布参数 $\theta$ 的估计（观测值 0,1,1）
**解题思路**：计算总体期望，令其等于样本均值。写出离散型似然函数求极值。
**解答**：
总体分布律：$P(X=1)=2\theta(1-\theta)$, $P(X=0)=\theta^2$, $P(X=2)=(1-\theta)^2$。样本观测值 $\{0, 1, 1\}$。
(1) **矩估计**：$E(X) = 1 \times 2\theta(1-\theta) + 2 \times (1-\theta)^2 = 2-2\theta$。样本均值 $\bar{x} = \frac{0+1+1}{3} = \frac{2}{3}$。令 $2-2\theta = 2/3$，解得 $\hat{\theta}_{1} = 2/3$。
(2) **最大似然估计**：似然函数 $L(\theta) = \theta^2 \cdot [2\theta(1-\theta)]^2 = 4\theta^4(1-\theta)^2$。
对数似然：$\ln L = \ln 4 + 4\ln\theta + 2\ln(1-\theta)$。求导 $\frac{4}{\theta} - \frac{2}{1-\theta} = 0$，解得 $\hat{\theta}_{2} = 2/3$。

#### 5. 均匀分布参数的矩估计（数据：1.3, 0.6, 1.7, 2.2, 1.1）
**解题思路**：均匀分布 $U(0, \beta)$ 的期望为 $\beta/2$，方差为 $\beta^2/12$。
**解答**：
样本均值 $\bar{x} = \frac{1.3+0.6+1.7+2.2+1.1}{5} = 1.38$。
(1) **总体均值估计**：$\hat{\mu} = \bar{x} = 1.38$。
(2) **总体方差估计**：$E(X) = \beta/2 = 1.38 \implies \hat{\beta} = 2.76$。方差估计 $\hat{\sigma}^2 = \hat{\beta}^2 / 12 = 2.76^2 / 12 \approx 0.6348$。
(3) **参数 $\beta$ 的矩估计**：$\hat{\beta} = 2\bar{x} = 2.76$。

#### 6. 无偏估计量与方差（分布律 1,2,3）
**解题思路**：根据无偏性 $E(T) = \theta$ 建立关于 $a_1, a_2, a_3$ 的方程，利用 $N_i$ 的期望求解。利用二项分布性质求方差。
**解答**：
$N_1, N_2, N_3$ 服从多项分布。$E(N_1) = n(1-\theta)$, $E(N_2) = n(\theta-\theta^2)$, $E(N_3) = n\theta^2$。
$T = a_1N_1 + a_2N_2 + a_3N_3$。
$E(T) = n[a_1(1-\theta) + a_2(\theta-\theta^2) + a_3\theta^2] = \theta$ 对一切 $\theta$ 成立。
比较系数：常数项为0 $\implies a_1=0$；一次项为 $1/n \implies a_2=1/n$；二次项为0 $\implies a_3=1/n$。
因此 $T = \frac{1}{n}(N_2+N_3) = \frac{1}{n}(n-N_1)$。
方差 $D(T) = \frac{1}{n^2}D(N_2+N_3) = \frac{1}{n^2}D(n-N_1) = \frac{1}{n^2}D(N_1)$。
由于 $N_1 \sim B(n, 1-\theta)$，$D(N_1) = n(1-\theta)\theta$。所以 $D(T) = \frac{\theta(1-\theta)}{n}$。

#### 7. 指数分布的参数估计
**解题思路**：利用期望等于 $\theta$ 进行矩估计。写出似然函数求导。
**解答**：
总体概率密度 $f(x) = \frac{1}{\theta}e^{-x/\theta}, x>0$，即 $X \sim Exp(1/\theta)$。
(1) **矩估计**：$E(X) = \theta$。令 $\theta = \bar{X}$，得 $\hat{\theta}_{1} = \bar{X}$。
(2) **最大似然估计**：$L(\theta) = \theta^{-n}e^{-\frac{1}{\theta}\sum X_i}$。对数似然 $\ln L = -n\ln\theta - \frac{1}{\theta}\sum X_i$。求导解得 $\hat{\theta}_{2} = \bar{X}$。

#### 8. 平移指数分布的最大似然估计
**解题思路**：似然函数单调递增，参数受样本最小值限制。
**解答**：
$f(x) = e^{-(x-\theta)}, x>\theta$。
似然函数 $L(\theta) = \prod e^{-(X_i-\theta)} = e^{-\sum X_i + n\theta}$，要求所有 $X_i > \theta$ 成立。
由于 $L(\theta)$ 关于 $\theta$ 单调递增，为使 $L(\theta)$ 最大，$\theta$ 应尽可能大，但必须小于所有样本值。因此 $\theta$ 的最大似然估计量为 $\hat{\theta} = \min(X_1, X_2, \dots, X_n)$。

#### 9. 几何分布的最大似然估计
**解题思路**：写出似然函数，求对数并求导。
**解答**：
$P(X=x) = p(1-p)^{x-1}, x=1,2,\dots$
似然函数 $L(p) = p^n(1-p)^{\sum X_i - n}$。对数似然 $\ln L = n\ln p + (\sum X_i - n)\ln(1-p)$。
求导并令为0：$\frac{n}{p} - \frac{\sum X_i - n}{1-p} = 0 \implies \hat{p} = \frac{n}{\sum X_i} = \frac{1}{\bar{X}}$。

#### 10. 最大次序统计量的分布与无偏估计
**解题思路**：先求分布函数，再求密度函数。计算期望令其等于 $\theta$ 求得 $c$。
**解答**：
$p(x, \theta) = \frac{3x^2}{\theta^3}, 0<x<\theta$。样本 $X_1, X_2, X_3$，令 $T = \max(X_1, X_2, X_3)$。
(1) 分布函数 $F_X(x) = \int_0^x \frac{3t^2}{\theta^3}dt = \frac{x^3}{\theta^3}$。$F_T(t) = [F_X(t)]^3 = \frac{t^9}{\theta^9}$。密度函数 $f_T(t) = \frac{9t^8}{\theta^9}, 0<t<\theta$。
(2) $E(T) = \int_0^\theta t \cdot \frac{9t^8}{\theta^9} dt = \frac{9}{10}\theta$。要使 $cT$ 是 $\theta$ 的无偏估计，需 $E(cT) = c \cdot \frac{9}{10}\theta = \theta \implies c = \frac{10}{9}$。

#### 11. 无偏估计量的有效性比较
**解题思路**：验证期望是否等于 $\mu$。计算方差，方差最小者最有效。
**解答**：
$E(X_i) = \mu, D(X_i) = \sigma^2$，且相互独立。
$E(\hat{\mu}_1) = \frac{1}{4}\mu + \frac{3}{4}\mu = \mu$，$E(\hat{\mu}_2) = \frac{1}{3}\mu + \frac{2}{3}\mu = \mu$，$E(\hat{\mu}_3) = \mu$，三者均为无偏估计。
方差：$D(\hat{\mu}_1) = \frac{1}{16}\sigma^2 + \frac{9}{16}\sigma^2 = \frac{10}{16}\sigma^2$；
$D(\hat{\mu}_2) = \frac{1}{9}\sigma^2 + \frac{4}{9}\sigma^2 = \frac{5}{9}\sigma^2 = \frac{10}{18}\sigma^2$；
$D(\hat{\mu}_3) = \frac{1}{4}\sigma^2 + \frac{1}{4}\sigma^2 = \frac{1}{2}\sigma^2 = \frac{9}{18}\sigma^2$。
因为 $D(\hat{\mu}_3) < D(\hat{\mu}_2) < D(\hat{\mu}_1)$，故 $\hat{\mu}_3$ 最有效。

#### 12. 组合无偏估计量使方差最小
**解题思路**：根据无偏性求 $a+b=1$，将 $b$ 替换为 $1-a$，对 $a$ 求导找极小值。
**解答**：
$E(\hat{\theta}_1) = E(\hat{\theta}_2) = \theta$，$D(\hat{\theta}_1) = 2D(\hat{\theta}_2)$。
$E(\hat{\theta}) = aE(\hat{\theta}_1) + bE(\hat{\theta}_2) = (a+b)\theta = \theta \implies a+b=1$。
方差 $D(\hat{\theta}) = a^2D(\hat{\theta}_1) + b^2D(\hat{\theta}_2) = 2a^2D(\hat{\theta}_2) + (1-a)^2D(\hat{\theta}_2) = (3a^2 - 2a + 1)D(\hat{\theta}_2)$。
对 $a$ 求导并令为0：$6a - 2 = 0 \implies a = \frac{1}{3}$。故 $b = \frac{2}{3}$。

#### 13. 泊松分布的无偏估计证明
**解题思路**：利用泊松分布期望和方差均为 $\lambda$，以及无偏估计定义。
**解答**：
总体 $X \sim P(\lambda)$，$E(X_i) = \lambda, D(X_i) = \lambda$。
(1) $E(\bar{X}) = \frac{1}{n}\sum E(X_i) = \lambda$，因此 $\bar{X}$ 是 $\lambda$ 的无偏估计量。
$E(S^2) = \sigma^2 = \lambda$，因此 $S^2$ 是 $\lambda$ 的无偏估计量。
(2) $E(\alpha\bar{X} + (1-\alpha)S^2) = \alpha E(\bar{X}) + (1-\alpha)E(S^2) = \alpha\lambda + (1-\alpha)\lambda = \lambda$。故也是无偏估计。

#### 14. 证明样本均值是总体均值的一致估计量
**解题思路**：利用大数定律（辛钦大数定律或切比雪夫不等式）。
**解答**：
由辛钦大数定律，设 $X_1, X_2, \dots, X_n$ 是独立同分布的随机变量序列，且 $E(X_i)=\mu$ 存在，则对于任意 $\epsilon > 0$，有：
$$ \lim_{n \to \infty} P\left( \left| \frac{1}{n}\sum_{i=1}^n X_i - \mu \right| < \epsilon \right) = 1 $$
即 $\bar{X} \xrightarrow{P} \mu$。因此样本均值 $\bar{X}$ 是总体均值 $\mu$ 的一致估计量。

## 1. 两点分布 $X\sim B(1,p)$

**思路**：矩估计用一阶原点矩；MLE构造似然函数。

- 总体期望：$(E(X)=p)$；样本一阶矩 $A_1=\bar X=\frac1n\sum X_i$
矩估计：$\hat p_1=\bar X$

似然函数：
$$
L(p)=\prod_{i=1}^n p^{X_i}(1-p)^{1-X_i}=p^{\sum X_i}(1-p)^{n-\sum X_i}
$$
$$
\ln L = \left(\sum X_i\right)\ln p + (n-\sum X_i)\ln(1-p)
$$
求导：
$$
\frac{d\ln L}{dp}=\frac{\sum X_i}{p}-\frac{n-\sum X_i}{1-p}=0 \Rightarrow \hat p_2=\bar X
$$
✅ **答案**：矩估计$\boldsymbol{\hat p_1=\bar X}$，最大似然估计$\boldsymbol{\hat p_2=\bar X}$

## 2. $f(x;\theta)=\begin{cases}(\theta+1)x^\theta,&0<x<1\0,&\text{其他}\end{cases},\theta>-1$

**思路**：一阶矩求矩估计；对数似然求MLE。
总体期望
$$
E(X)=\int_0^1 x\cdot (\theta+1)x^\theta dx=(\theta+1)\int_0^1 x^{\theta+1}dx=\frac{\theta+1}{\theta+2}
$$
令 $E(X)=\bar X$
$$
\bar X=\frac{\theta+1}{\theta+2}\Rightarrow \hat\theta_1=\frac{2\bar X-1}{1-\bar X}
$$

似然函数：
$$
L(\theta)=\prod_{i=1}^n (\theta+1)x_i^\theta=(\theta+1)^n\left(\prod x_i\right)^\theta
$$
$$
\ln L=n\ln(\theta+1)+\theta\sum\ln x_i
$$
$$
\frac{d\ln L}{d\theta}=\frac{n}{\theta+1}+\sum\ln x_i=0
$$
$$
\boldsymbol{\hat\theta_2=-1-\frac{n}{\sum_{i=1}^n\ln X_i}}
$$

✅ **答案**：矩估计$\boldsymbol{\hat\theta_1=\dfrac{2\bar X-1}{1-\bar X}}$，MLE$\boldsymbol{\hat\theta_2=-1-\dfrac{n}{\sum\ln X_i}}$

## 4. 分布列

| $X$ | 1 | 0 | 2 |
| --- | --- | --- | --- |
| $P$ | $2\theta(1-\theta)$ | $\theta^2$ | $(1-\theta)^2$ |

### (1) 矩估计

$$
E(X)=1\cdot 2\theta(1-\theta)+0\cdot\theta^2+2\cdot(1-\theta)^2=2\theta-2\theta^2+2(1-2\theta+\theta^2)=2-2\theta

$$

### (2) 最大似然估计

$$

$$
\ln L=\ln4+4\ln\theta+2\ln(1-\theta)

$$
\frac{d\ln L}{d\theta}=\frac4\theta-\frac{2}{1-\theta}=0 \Rightarrow 4(1-\theta)=2\theta \Rightarrow 4-4\theta=2\theta \Rightarrow \hat\theta=\frac23
$$

## 6. 分布列

| $X$ | 1 | 2 | 3 |
| --- | --- | --- | --- |
| $P$ | $1-\theta$ | $\theta-\theta^2$ | $\theta^2$ |
| $N_i$：样本中取值$i$的个数，$n=N_1+N_2+N_3$，$T=\sum_{i=1}^3 a_i N_i$，要求$E(T)=\theta$。 |  |  |  |

$$
E(N_1)=n(1-\theta),\ E(N_2)=n(\theta-\theta^2),\ E(N_3)=n\theta^2
$$
$$
E(T)=a_1n(1-\theta)+a_2n(\theta-\theta^2)+a_3n\theta^2=\theta
$$
整理：
$$
a_1n + n(-a_1+a_2)\theta + n(-a_2+a_3)\theta^2=\theta
$$
比较多项式系数：
$$
\begin{cases}
a_1 n =0\
n(-a_1+a_2)=1\
n(-a_2+a_3)=0
\end{cases}
$$
$n>0\Rightarrow a_1=0,\ a_2=\frac1n,\ a_3=\frac1n$。
$$
T=\frac1n N_2+\frac1n N_3=\frac{N_2+N_3}{n}
$$
求方差：
$N_1,N_2,N_3$服从**多项分布**。
$$
D(N_2+N_3)=D(n-N_1)=D(N_1)=n(1-\theta)\theta
$$
$$
D(T)=\frac{1}{n^2}D(N_2+N_3)=\frac{\theta(1-\theta)}{n}
$$

✅ **答案**：$\boldsymbol{a_1=0,a_2=a_3=\dfrac1n}$；$\boldsymbol{D(T)=\dfrac{\theta(1-\theta)}{n}}$

##7. 指数分布 $f(x;\theta)=\begin{cases}\dfrac1\theta e^{-x/\theta},&x>0\0,&x\le0\end{cases},\theta>0$
**思路**：指数分布$E(X)=\theta$。
矩估计：$\bar X=\theta \Rightarrow \boldsymbol{\hat\theta_1=\bar X}$

似然函数：
$$
L(\theta)=\prod_{i=1}^n\frac1\theta e^{-x_i/\theta}=\frac1{\theta^n}\exp\left(-\frac1\theta\sum x_i\right)
$$
$$
\ln L=-n\ln\theta-\frac1\theta\sum x_i
$$
$$
\frac{d\ln L}{d\theta}=-\frac{n}{\theta}+\frac{\sum x_i}{\theta^2}=0 \Rightarrow \boldsymbol{\hat\theta_2=\bar X}
$$

✅ **答案**：矩估计$\boldsymbol{\hat\theta_1=\bar X}$，MLE$\boldsymbol{\hat\theta_2=\bar X}$

##8. $f(x;\theta)=\begin{cases}e^{-(x-\theta)},&x>\theta\0,&x\le\theta\end{cases},\theta>0$

##9. 几何分布 $P(X=x)=p(1-p)^{x-1},\ x=1,2,\dots$
