### 思考题/习题解答

#### 1~4题（填空与选择）

**1. 假设检验可分为两类，它们是______和______。**
*   **解题思路**：根据是否对总体分布做出特定假设，假设检验分为参数检验和非参数检验。
*   **解答**：参数检验；非参数检验。

**2. 在假设检验中，用 $\alpha$ 和 $\beta$ 分别表示犯第一类错误和第二类错误的概率，则当样本容量一定时，下列说法正确的是( )。**
*   **解题思路**：在样本量固定的情况下，$\alpha$ 和 $\beta$ 不能同时减小。减小 $\alpha$ 会导致 $\beta$ 增大，反之亦然。要同时减小两者，必须增大样本容量 $n$。
*   **解答**：(C)。$\alpha$ 与 $\beta$ 不能同时减小，减小其中一个，另一个往往就会增大。

**3. 设 $X_1, X_2, \dots, X_n$ 是来自正态总体 $X \sim N(\mu, \sigma^2)$ 的样本，$\sigma^2$ 为已知常数，要检验假设 $H_0: \mu = \mu_0$（$\mu_0$ 为已知常数），应用______检验法；检验的统计量是______；当 $H_0$ 成立时，该统计量服从______分布。**
*   **解题思路**：方差已知的正态总体均值检验，使用U检验法（Z检验法）。检验统计量为 $Z = \frac{\bar{X} - \mu_0}{\sigma / \sqrt{n}}$，在 $H_0$ 成立下服从标准正态分布。
*   **解答**：U（或Z）；$$Z = \frac{\bar{X} - \mu_0}{\sigma / \sqrt{n}}$$；标准正态分布 $N(0,1)$。

**4. 总体 $X \sim N(\mu, \sigma^2)$，$\mu$ 未知，$X_1, X_2, \dots, X_n$ 为来自总体 $X$ 的样本，记 $\bar{X}$ 为样本均值，$S^2$ 为样本方差。要检验假设 $H_0: \sigma^2 \ge 2$，$H_1: \sigma^2 < 2$，应取检验统计量 $\chi^2$ 为( )。**
*   **解题思路**：方差检验的统计量为 $\chi^2 = \frac{(n-1)S^2}{\sigma_0^2}$，在 $H_0$ 的边界 $\sigma^2 = 2$ 下，统计量为 $\frac{(n-1)S^2}{2}$。
*   **解答**：(D)。$\frac{(n-1)S^2}{2}$。

#### 5~17题（计算题）

**5. 设从 $\sigma = 10$ 的正态总体中随机抽取 $n = 25$ 的样本，要以 $\alpha = 0.05$ 的水平拒绝 $H_0: \mu = 100$，问 $\bar{x}$ 要多大？若 $n = 100$，要以 $\alpha = 0.01$ 的水平拒绝 $H_0: \mu = 140$，求其拒绝域？**
*   **解题思路**：
    已知 $\sigma$，采用Z检验。$H_0: \mu = \mu_0$，$H_1: \mu \neq \mu_0$（无方向暗示，取双侧）。
    统计量 $$Z = \frac{\bar{X} - \mu_0}{\sigma / \sqrt{n}} \sim N(0,1)$$。
    拒绝域为 $|Z| \ge z_{\alpha/2}$。
*   **解答**：
    (1) 当 $n=25, \sigma=10, \mu_0=100, \alpha=0.05$ 时，$z_{0.025} = 1.96$。
    $$\left| \frac{\bar{x} - 100}{10 / \sqrt{25}} \right| \ge 1.96 \implies \left| \frac{\bar{x} - 100}{2} \right| \ge 1.96$$
    $$|\bar{x} - 100| \ge 3.92 \implies \bar{x} \ge 103.92 \text{ 或 } \bar{x} \le 96.08$$
    因此，$\bar{x}$ 要大于等于 103.92 或小于等于 96.08 才能拒绝 $H_0$。
    (2) 当 $n=100, \sigma=10, \mu_0=140, \alpha=0.01$ 时，$z_{0.005} = 2.58$。
    $$\left| \frac{\bar{x} - 140}{10 / \sqrt{100}} \right| \ge 2.58 \implies \left| \frac{\bar{x} - 140}{1} \right| \ge 2.58$$
    故拒绝域为：$\bar{x} \ge 142.58$ 或 $\bar{x} \le 137.42$。

**6. 设某次考试学生的成绩服从正态分布，从中随机地抽取36位考生的成绩，算得平均成绩为66.5分，标准差为15分，问在显著性水平0.05下，是否可以认为这次考试全体考生的平均成绩为70分？**
*   **解题思路**：
    总体正态，方差未知，用t检验。$H_0: \mu = 70$，$H_1: \mu \neq 70$。
    统计量 $$t = \frac{\bar{X} - \mu_0}{S / \sqrt{n}} \sim t(n-1)$$。
*   **解答**：
    $\mu_0 = 70, n = 36, \bar{x} = 66.5, s = 15, \alpha = 0.05$。
    $$t = \frac{66.5 - 70}{15 / \sqrt{36}} = \frac{-3.5}{2.5} = -1.4$$
    自由度 $df = 35$，查t分布表得双侧临界值 $t_{0.025}(35) \approx 2.0301$。
    因为 $|t| = 1.4 < 2.0301$，所以不能拒绝 $H_0$。
    **结论**：在显著性水平0.05下，可以认为这次考试全体考生的平均成绩为70分。

**7. 某企业的质量检验规定，整批产品关于某项指标的平均值达60以上才可以出厂，若在一批产品中重复抽取50件对该项指标进行检测，计算得样本均值 $\bar{x} = 64$，标准差 $s = 9$，试问：（$\alpha = 0.05$）（1）这批产品关于该项指标的均值是否与60有显著差异？（2）这批产品能否出厂？**
*   **解题思路**：
    (1) 双侧检验 $H_0: \mu = 60$，$H_1: \mu \neq 60$。
    (2) 单侧检验 $H_0: \mu \le 60$，$H_1: \mu > 60$。采用t检验。
*   **解答**：
    $n = 50, \bar{x} = 64, s = 9, \alpha = 0.05$。
    (1) $$t = \frac{64 - 60}{9 / \sqrt{50}} = \frac{4}{1.2728} \approx 3.143$$
    查t分布表，$t_{0.025}(49) \approx 2.01$。$|t| = 3.143 > 2.01$，拒绝 $H_0$。
    **结论**：这批产品关于该项指标的均值与60有显著差异。
    (2) 建立单侧检验：$H_0: \mu \le 60$，$H_1: \mu > 60$。
    临界值 $t_{0.05}(49) \approx 1.677$。由于 $t = 3.143 > 1.677$，拒绝 $H_0$。
    **结论**：在0.05的显著性水平下，可以认为平均指标显著大于60，这批产品能出厂。

**8. 某种羊毛在处理前后各抽取样本测得含脂率如下：处理前：0.19, 0.18, 0.30, 0.66, 0.42, 0.08, 0.12, 0.30, 0.27, 0.21；处理后：0.15, 0.13, 0.07, 0.24, 0.19, 0.04, 0.08, 0.20。问经过处理后含脂率（假定含脂率服从正态分布，等方差）有无显著变化（$\alpha = 0.05$）？**
*   **解题思路**：
    两独立样本t检验（等方差）。
    $H_0: \mu_1 = \mu_2$，$H_1: \mu_1 \neq \mu_2$。
    统计量 $$t = \frac{\bar{X}_1 - \bar{X}_2}{S_p \sqrt{\frac{1}{n_1} + \frac{1}{n_2}}} \sim t(n_1+n_2-2)$$
    其中合并方差 $$S_p^2 = \frac{(n_1-1)S_1^2 + (n_2-1)S_2^2}{n_1+n_2-2}$$。
*   **解答**：
    $n_1 = 10$，计算得 $\bar{x}_1 = 0.273$，$s_1^2 \approx 0.0283$。
    $n_2 = 8$，计算得 $\bar{x}_2 = 0.1375$，$s_2^2 \approx 0.0047$。
    $$S_p^2 = \frac{9 \times 0.0283 + 7 \times 0.0047}{16} \approx 0.0180 \implies S_p \approx 0.1342$$
    $$t = \frac{0.273 - 0.1375}{0.1342 \times \sqrt{\frac{1}{10} + \frac{1}{8}}} \approx \frac{0.1355}{0.1342 \times 0.4743} \approx 2.129$$
    自由度 $df = 16$，查表得 $t_{0.025}(16) = 2.120$。
    因为 $|t| = 2.129 > 2.120$，拒绝 $H_0$。
    **结论**：经过处理后含脂率有显著变化。

**9. 某车间生产的金属丝，质量较为稳定，折断力方差 $\sigma_0^2 = 64$，今从一批产品中抽10根作折断力试验，结果为：578, 572, 570, 568, 572, 570, 572, 596, 584, 570。假定折断力服从正态分布，问是否可以相信这批金属丝的折断力方差仍是64（$\alpha = 0.05$）？**
*   **解题思路**：
    单样本方差检验。$H_0: \sigma^2 = 64$，$H_1: \sigma^2 \neq 64$。
    统计量 $$\chi^2 = \frac{(n-1)S^2}{\sigma_0^2} \sim \chi^2(n-1)$$。
*   **解答**：
    $n = 10$，计算样本均值 $\bar{x} = 575.2$。

# 习题7 全题解答（假设检验）

> 
> 核心知识点回顾
> 
> 
> 1. **第一类错误（弃真错误）**：$H_0$为真，拒绝$H_0$，概率$\alpha$；**第二类错误（取伪错误）**：$H_0$为假，接受$H_0$，概率$\beta$。样本容量固定时，$\alpha\uparrow\Rightarrow\beta\downarrow$，不能同时减小。
> 2. **Z检验（u检验）**：正态总体，$\sigma^2$已知，检验均值$\mu$；统计量
> $$Z=\frac{\overline{X}-\mu_0}{\sigma/\sqrt{n}}\sim (N(0,1))$$
> 3. **t检验**：正态总体，$\sigma^2$未知，检验均值$\mu$；统计量
> $$t=\frac{\overline{X}-\mu_0}{S/\sqrt{n}}\sim t(n-1)$$
> 4. **$\boldsymbol{\chi^2}$卡方检验**：正态总体，$\mu$未知，检验方差$\sigma^2$；统计量
> $$\chi^2=\frac{(n-1)S^2}{\sigma_0^2}\sim \chi^2(n-1)$$
> 5. 大样本下，用中心极限定理近似Z检验；**$\chi^2$拟合优度检验**检验总体分布；**列联表$\boldsymbol{\chi^2}$独立性检验**检验两个分类变量是否独立；**秩和检验**是非参数检验。

> 思路：假设检验分为两大类：
