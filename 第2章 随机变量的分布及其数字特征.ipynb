## 习题1
一袋中装有 7 只白球，3 只黑球。每次任取一个不放回，以 $X$ 表示首次取到白球时取球次数。试求随机变量 $X$ 的概率分布列、分布函数。

解题思路：不放回抽取，属于超几何分布模型。$X$ 的取值为1,2,3,4。利用组合数公式求概率分布列，再根据分布函数定义 $F(x) = P(X \le x)$ 求分段函数。

解答：

离散不放回，$X$ 可取 $1,2,3,4$

分布律$P(X=k)$=前$k-1$黑，第$k$白

分布函数$F(x)=\sum_{x_i\le x}P(X=x_i)$
- $P(X=1)=\dfrac7{10}$
- $P(X=2)=\dfrac{3}{10}\cdot\dfrac79=\dfrac{7}{30}$
- $P(X=3)=\dfrac{3}{10}\cdot\dfrac29\cdot\dfrac78=\dfrac{7}{120}$
- $P(X=4)=\dfrac{3}{10}\cdot\dfrac29\cdot\dfrac18\cdot\dfrac77=\dfrac{1}{120}$

**分布律**
|$X$|1|2|3|4|
|---|---|---|---|---|
|$P$|$\dfrac7{10}$|$\dfrac{7}{30}$|$\dfrac{7}{120}$|$\dfrac{1}{120}$|


**分布函数**
$$
F(x)=
\begin{cases}
0, &x<1\\
\dfrac{7}{10}, &1\le x<2\\
\dfrac{7}{10}+\dfrac{7}{30}=\frac{28}{30}, &2\le x<3\\
\dfrac{28}{30}+\dfrac{7}{120}=\frac{119}{120}, &3\le x<4\\
1, &x\ge4
\end{cases}
$$


## 习题2

设离散型随机变量 $X$ 的概率分布列为
$$P\{X=k\} = ae^{-k}, \quad k=1,2,\cdots,$$

试确定常数 $a$。

解题思路：利用离散型随机变量概率分布列的归一性，即 $\sum\limits_{k=1}^{\infty} P\{X=k\} = 1$，结合等比级数求和公式求解。

解答：
由概率分布列的性质可知：

$$\sum_{k=1}^{\infty} P\{X=k\} = \sum_{k=1}^{\infty} a e^{-k} = a \sum_{k=1}^{\infty} (e^{-1})^k = a \frac{e^{-1}}{1 - e^{-1}} = 1$$

## 习题3

电子线路中装有两个并联的继电器，假设这两个继电器是否接通具有随机性，且彼此独立，已知每个继电器接通的概率为 0.8，记 $X$ 为线路中接通的继电器的个数。求：(1) $X$ 的概率分布列；(2) 线路接通的概率。

解题思路：两个继电器独立工作，$X$ 为接通的个数，服从二项分布 $B(2, 0.8)$。线路接通等价于至少有一个继电器接通，即 $P(X \ge 1) = 1 - P(X=0)$。

解答：

(1) $X \sim B(2, 0.8)$，其概率分布列为：
$$P\{X=k\} = C_2^k (0.8)^k (0.2)^{2-k}, \quad k=0,1,2$$
$$P\{X=0\} = 0.2^2 = 0.04$$
$$P\{X=1\} = C_2^1 (0.8)(0.2) = 0.32$$
$$P\{X=2\} = 0.8^2 = 0.64$$

分布律
|$X$|$0$|$1$|$2$|
|---|---|---|---|
|$P$|$0.04$|$0.32$|$0.64$|

(2) 线路接通的概率为：
$$P\{X \ge 1\} = P\{X=1\} + P\{X=2\} = 0.32 + 0.64 = 0.96$$
或 $1 - P\{X=0\} = 1 - 0.04 = 0.96$

## 习题4

某出租汽车公司共有出租车 400 辆，设每天每辆出租车出现故障的概率为 0.02，试求一天内没有出租车出现故障的概率。

解题思路：设出现故障的出租车数为 $X$，$X \sim B(400, 0.02)$。因为 $n$ 很大，$p$ 很小，可以用泊松定理近似计算，$\lambda = np = 400 \times 0.02 = 8$。

解答：

设一天内出现故障的出租车数为 $X$，则 $X \sim B(400, 0.02)$。

$n$ 大 $p$ 小，泊松近似 $\lambda=np=400\times0.02=8$，泊松 $P(X=k)=\dfrac{\lambda^k e^{-\lambda}}{k!}$

没有出租车出现故障的概率为 $P\{X=0\}$：
$$P\{X=0\} = C_{400}^0 (0.02)^0 (0.98)^{400} \approx \frac{\lambda^0 e^{-\lambda}}{0!} = e^{-8} \approx 0.000335$$

## 习题6

设连续型随机变量 $X$ 的分布函数为
$$F(x) = \begin{cases}
a+be^{-\frac{x^2}{2}}, & x \ge 0, \\
0, & x < 0,
\end{cases}$$

(1) 求常数 $a$ 和 $b$；(2) 求随机变量 $X$ 的概率密度函数；(3) 求 $P\{\sqrt{\ln 4} < X \le \sqrt{\ln 16}\}$。

解题思路：利用分布函数的性质：$F(+\infty)=1$ 以及连续型随机变量分布函数在分界点处的连续性 $F(0^+)=F(0)$ 来求常数 $a, b$。概率密度函数 $f(x) = F'(x)$。区间概率 $P\{a < X \le b\} = F(b) - F(a)$。

解答：

(1) 由分布函数性质 $\lim_{x \to +\infty} F(x) = 1$，得 $a = 1$。

又因为 $F(x)$ 在 $x=0$ 处连续，$F(0) = a + b e^0 = a + b = 0$，所以 $b = -1$。

(2) 随机变量 $X$ 的概率密度函数为：
$$f(x) = F'(x) = \begin{cases} x e^{-x^2/2}, & x \ge 0, \\ 0, & x < 0. \end{cases}$$

(3) 
$$P\{\sqrt{\ln 4} < X \le \sqrt{\ln 16}\} = F(\sqrt{\ln 16}) - F(\sqrt{\ln 4})$$
$$= (1 - e^{-\frac{\ln 16}{2}}) - (1 - e^{-\frac{\ln 4}{2}}) = (1 - e^{-\ln 4}) - (1 - e^{-\ln 2}) = (1 - \frac{1}{4}) - (1 - \frac{1}{2}) = \frac{3}{4} - \frac{1}{2} = \frac{1}{4}$$

## ❗习题7

设某地区每天的用电量 $X$ (单位:$1 \times 10^4 \text{ kW}\cdot\text{h}$) 是一连续型随机变量，其概率密度函数为
$$f(x) = \begin{cases}
12x(1-x)^2, & 0<x<1, \\
0, & \text{其他}.
\end{cases}$$

假设该地区每天的供电量仅有 $8 \times 10^4 \text{ kW}\cdot\text{h}$，求该地区每天供电量不足的概率。若每天的供电量上升到 $9 \times 10^4 \text{ kW}\cdot\text{h}$，每天供电量不足的概率又是多少？

解题思路：供电不足即用电量 $X$ 大于供电量。需要计算 $P\{X > 0.8\}$ 和 $P\{X > 0.9\}$。（注：题目中 $8\times10^4$ 和 $9\times10^4$ 应为 $0.8\times10^4$ 和 $0.9\times10^4$，否则概率为0）。

解答：

设供电量为 $0.8 \times 10^4 \text{ kW}\cdot\text{h}$ 时，供电不足的概率为：
$$P\{X > 0.8\} = \int_{0.8}^1 12x(1-x)^2 dx = \int_{0.8}^1 12(x^3 - 2x^2 + x) dx = 12 \left[ \frac{x^4}{4} - \frac{2x^3}{3} + \frac{x^2}{2} \right]_{0.8}^1$$
$$= 12 \left[ \left( \frac{1}{4} - \frac{2}{3} + \frac{1}{2} \right) - \left( \frac{0.8^4}{4} - \frac{2 \times 0.8^3}{3} + \frac{0.8^2}{2} \right) \right] = 1 - 12(0.0256 - 0.3413 + 0.32) = 0.9488$$
若供电量上升到 $0.9 \times 10^4 \text{ kW}\cdot\text{h}$：
$$P\{X > 0.9\} = \int_{0.9}^1 12x(1-x)^2 dx = 12 \left[ \frac{x^4}{4} - \frac{2x^3}{3} + \frac{x^2}{2} \right]_{0.9}^1 = 1 - 12(0.1640 - 0.486 + 0.405) = 0.0037$$

用电量$X$，密度$f(x)=\begin{cases}12x(1-x)^2, &0<x<1\\0,&\text{其他}\end{cases}$，单位$1\times10^8\mathrm{kW\cdot h}$
供电$8\times10^7$即$\(x=0\).8$；供电$9\times10^7$即$\(x=0\).9$；供电不足$\iff X>x_0$。
**思路**：连续概率$P(X>x_0)=\int_{x_0}^1 f(x)dx$。
$$
\begin{align*}
P(X>0.8)&=\int_{0.8}^1 12x(1-x)^2 dx=\int_{0.8}^1 12(x-2x^2+x^3)dx\\
&=12\left[\frac{x^2}{2}-\frac{2x^3}{3}+\frac{x^4}{4}\right]_{0.8}^1=\(\boldsymbol{0.0272}\)\\
P(X>0.9)&=\int_{0.9}^112x(1-x)^2d\(x=0\).0037
\end{align*}
$$
供电$8\times10^7$不足概率$\boldsymbol{\(\boldsymbol{0.0272}\)}$；供电$9\times10^7$不足概率$\boldsymbol{\(\boldsymbol{0.0037}\)}$


## 习题8

一座大楼装有 5 个同类型的供水设备。调查表明在任一时刻 $t$ 每个设备被使用的概率为 0.1，问在同一时刻，(1) 恰有 2 个设备被使用的概率是多少？(2) 至少有 3 个设备被使用的概率是多少？(3) 至多有 3 个设备被使用的概率是多少？(4) 至少有 1 个设备被使用的概率是多少？

解题思路：5个设备独立工作，被使用的个数 $X$ 服从二项分布 $X \sim B(5, 0.1)$。利用二项分布概率公式计算各问。

解答：

(1) 恰有2个设备被使用的概率：
$$P\{X=2\} = C_5^2 (0.1)^2 (0.9)^3 = 10 \times 0.01 \times 0.729 = 0.0729$$

(2) 至少有3个设备被使用的概率：
$$P\{X \ge 3\} = P\{X=3\} + P\{X=4\} + P\{X=5\} = C_5^3 (0.1)^3 (0.9)^2 + C_5^4 (0.1)^4 (0.9) + C_5^5 (0.1)^5 = 0.00856$$

(3) 至多有3个设备被使用的概率：
$$P\{X \le 3\} = 1 - P\{X=4\} - P\{X=5\} = 1 - 0.00045 - 0.00001 = 0.99954$$

(4) 至少有1个设备被使用的概率：
$$P\{X \ge 1\} = 1 - P\{X=0\} = 1 - (0.9)^5 = 1 - 0.59049 = 0.40951$$

## 习题9

设随机变量 $X \sim U(-2,2)$，求方程 $y^2+3Xy+4=0$ 有实根的概率。

解题思路：一元二次方程 $y^2 + 3Xy + 4 = 0$ 有实根的条件是判别式 $\Delta = (3X)^2 - 16 \ge 0$，即 $|X| \ge 4/3$。由于 $X \sim U(-2,2)$，概率密度为 $f(x) = 1/4, x \in (-2,2)$，通过积分求概率。

解答：

方程有实根需满足 $\Delta = 9X^2 - 16 \ge 0 \implies X^2 \ge 16/9 \implies X \ge 4/3$ 或 $X \le -4/3$。
$$P = P\{X \ge \frac{4}{3}\} + P\{X \le -\frac{4}{3}\} = \int_{4/3}^2 \frac{1}{4} dx + \int_{-2}^{-4/3} \frac{1}{4} dx = \frac{1}{4} \left( 2 - \frac{4}{3} \right) + \frac{1}{4} \left( -\frac{4}{3} + 2 \right) = \frac{1}{6} + \frac{1}{6} = \frac{1}{3}$$

## 习题10

已知顾客在某银行窗口等待服务的时间 $X$ (单位:min) 服从指数分布，其概率密度函数为
$$f(x) = \begin{cases}
\frac{1}{5}e^{-\frac{x}{5}}, & x>0, \\
0, & x \le 0.
\end{cases}$$
一顾客在窗口等待服务，若超过 10 min，他就离开，他一个月要到银行 5 次，求他一个月内未等到服务而离开窗口的次数 $Y$ 服从的概率分布，并求 $P\{Y \ge 1\}$。

解题思路：等待时间 $X$ 服从指数分布。先求顾客未等到服务而离开的概率 $p = P\{X > 10\}$。一个月去5次，离开次数 $Y \sim B(5, p)$。

解答：

顾客未等到服务而离开的概率为：
$$p = P\{X > 10\} = \int_{10}^{\infty} \frac{1}{5} e^{-x/5} dx = e^{-10/5} = e^{-2}$$
$Y$ 服从二项分布 $B(5, e^{-2})$，概率分布为：
$$P\{Y=k\} = C_5^k (e^{-2})^k (1-e^{-2})^{5-k}, \quad k=0,1,2,3,4,5$$
$$P\{Y \ge 1\} = 1 - P\{Y=0\} = 1 - (1-e^{-2})^5$$

## 习题11

已知 $X \sim N(0,1)$，求 $P\{X>0\}, P\{|X|<3\}, P\{0<X \le 3\}, P\{X>3\}$。

解题思路：标准正态分布 $X \sim N(0,1)$，利用标准正态分布表 $\Phi(x)$ 计算。注意对称性 $\Phi(-x) = 1 - \Phi(x)$。

解答：
$$P\{X > 0\} = 1 - \Phi(0) = 0.5$$
$$P\{|X| < 3\} = P\{-3 < X < 3\} = \Phi(3) - \Phi(-3) = 2\Phi(3) - 1$$
$$P\{0 < X \le 3\} = \Phi(3) - \Phi(0) = \Phi(3) - 0.5$$
$$P\{X > 3\} = 1 - \Phi(3)$$
（具体数值需查标准正态分布表，如 $\Phi(3) \approx 0.9987$）

## 习题12

已知 $X \sim N(1,2^2)$，求 $P\{X<-3\}, P\{1<X \le 3\}, P\{|X|<3\}$。

解题思路：一般正态分布标准化。$X \sim N(1, 2^2)$，令 $Z = \frac{X-1}{2} \sim N(0,1)$。

解答：
$$P\{X < -3\} = P\left\{ \frac{X-1}{2} < \frac{-3-1}{2} \right\} = P\{Z < -2\} = \Phi(-2) = 1 - \Phi(2)$$
$$P\{1 < X \le 3\} = P\left\{ \frac{1-1}{2} < Z \le \frac{3-1}{2} \right\} = P\{0 < Z \le 1\} = \Phi(1) - \Phi(0) = \Phi(1) - 0.5$$
$$P\{|X| < 3\} = P\{-3 < X < 3\} = P\left\{ \frac{-3-1}{2} < Z < \frac{3-1}{2} \right\} = \Phi(1) - \Phi(-2) = \Phi(1) + \Phi(2) - 1$$

## 习题14

设随机变量 $X \sim N(0,1)$，求下列随机变量 $Y$ 的概率密度函数：

(1) $Y=2X-1$；(2) $Y=e^{-X}$；(3) $Y=X^2$。

**14. 解题思路**：连续型随机变量函数的分布。利用公式法 $f_Y(y) = f_X(h(y)) |h'(y)|$ 或分布函数法。
**解答**：
(1) $Y = 2X - 1$。$X = \frac{Y+1}{2}$，$\frac{dx}{dy} = \frac{1}{2}$。
$$f_Y(y) = f_X\left( \frac{y+1}{2} \right) \left| \frac{1}{2} \right| = \frac{1}{2\sqrt{2\pi}} e^{-\frac{(y+1)^2}{8}}, \quad -\infty < y < \infty$$
即 $Y \sim N(-1, 4)$。
(2) $Y = e^{-X}$。$y > 0$ 时，$X = -\ln y$，$\left| \frac{dx}{dy} \right| = \frac{1}{y}$。
$$f_Y(y) = f_X(-\ln y) \frac{1}{y} = \frac{1}{y\sqrt{2\pi}} e^{-\frac{(\ln y)^2}{2}}, \quad y > 0$$
(3) $Y = X^2$。$y > 0$ 时，
$$f_Y(y) = f_X(\sqrt{y}) \frac{1}{2\sqrt{y}} + f_X(-\sqrt{y}) \frac{1}{2\sqrt{y}} = \frac{2}{2\sqrt{2\pi y}} e^{-y/2} = \frac{1}{\sqrt{2\pi y}} e^{-y/2}, \quad y > 0$$


$X\sim N(0,1)$，求下面$Y$的概率密度。
**思路**：随机变量单调变换公式，若$\(y=\(g(x)\)\)$严格单调，$f_Y(y)=f_X(g^{-1}(y))\left|\frac{d}{dy}g^{-1}(y)\right|$，$f_X(x)=\frac1{\sqrt{2\pi}}e^{-x^2/2}$

(1)$Y=2X-1$，线性变换，正态仍正态，$Y\sim N(-1,4)$
$$f_Y(y)=\frac1{2\sqrt{2\pi}}\exp\left(-\frac{(y+1)^2}{8}\right),\quad -\infty<y<+\infty$$

(2)$Y=e^{-X}$，$\(y>0\)$，$x=-\ln y,\frac{dx}{dy}=-\frac1y$
$$
f_Y(y)=
\begin{cases}
\frac1{\sqrt{2\pi}y}\exp\left(-\frac{(\ln y)^2}{2}\right), &\(y>0\)\\
0, &y\le0
\end{cases}
$$

(3)$Y=X^2$，卡方$\chi^2(1)$
$$
f_Y(y)=
\begin{cases}
\frac1{\sqrt{2\pi y}}e^{-y/2}, &\(y>0\)\\
0, &y\le0
\end{cases}
$$


## 习题15

设随机变量 $X \sim U(0,\pi)$，求随机变量 $Y=2\ln X$ 的概率密度函数。

**15. 解题思路**：$X \sim U(0, \pi)$，$f_X(x) = 1/\pi$。$Y = 2\ln X$ 单调，值域 $y \in (-\infty, 2\ln \pi)$。
**解答**：
$y = 2\ln x \implies x = e^{y/2}$，$\frac{dx}{dy} = \frac{1}{2} e^{y/2}$。
$$f_Y(y) = f_X(e^{y/2}) \left| \frac{1}{2} e^{y/2} \right| = \frac{1}{\pi} \cdot \frac{1}{2} e^{y/2} = \frac{1}{2\pi} e^{y/2}, \quad y < 2\ln \pi$$


$X\sim U(0,\pi)$，$f_X(x)=\frac1\pi,0<x<\pi$；$Y=2\ln X$
**思路**：单调变换。$y=2\ln x\Rightarrow x=e^{y/2},\frac{dx}{dy}=\frac12 e^{y/2}$；$0<x<\pi\Rightarrow -\infty<y<2\ln\pi$
$$
f_Y(y)=f_X(e^{y/2})\cdot\left|\frac{dx}{dy}\right|
=\frac1\pi\cdot\frac12 e^{y/2},\quad -\infty<y<2\ln\pi
$$
$$
f_Y(y)=
\begin{cases}
\frac{1}{2\pi}e^{y/2}, &y<2\ln\pi\\
0, &y\ge2\ln\pi
\end{cases}
$$

## 习题17

设随机变量 $X$ 的概率函数为
$$P\{X=k\} = \frac{a^k}{(1+a)^{k+1}}, \quad k=0,1,2,\cdots,$$
其中 $a>0$ 是常数，求 $E(X)$。

**17. 解题思路**：利用级数求和公式求数学期望。该分布为参数 $p = \frac{1}{1+a}$ 的几何分布（取值从0开始）。
**解答**：
$$E(X) = \sum_{k=0}^{\infty} k \frac{a^k}{(1+a)^{k+1}} = \frac{1}{1+a} \sum_{k=1}^{\infty} k \left( \frac{a}{1+a} \right)^k$$
利用级数公式 $\sum_{k=1}^{\infty} k x^k = \frac{x}{(1-x)^2}$（$|x|<1$），令 $x = \frac{a}{1+a}$：
$$E(X) = \frac{1}{1+a} \cdot \frac{\frac{a}{1+a}}{\left( 1 - \frac{a}{1+a} \right)^2} = \frac{1}{1+a} \cdot \frac{\frac{a}{1+a}}{\left( \frac{1}{1+a} \right)^2} = a$$

$$\(P(X=k)\)=\frac{a^k}{(1+a)^{k+1}},\ k=0,1,2\cdots,\ a>0$$求$\(E(X)\)$
**思路**：离散期望定义，令$q=\frac{a}{1+a}$，$\(P(X=k)\)=\frac{1}{1+a}\cdot q^k$；级数$\sum_{k=0}^{\infty}k q^k=\frac{q}{(1-q)^2}$
$$
\(E(X)\)=\sum_{k=0}^{\infty}k\cdot \frac{a^k}{(1+a)^{k+1}}=\frac{1}{1+a}\sum_{k=1}^{\infty}k\left(\frac{a}{1+a}\right)^k
$$
$q=\frac{a}{1+a},1-q=\frac{1}{1+a}$
$$
\(E(X)\)=\frac{1}{1+a}\cdot\frac{q}{(1-q)^2}=\frac{1}{1+a}\cdot \frac{\frac{a}{1+a}}{\left(\frac{1}{1+a}\right)^2}=\boldsymbol{a}
$$


## 习题18

设随机变量 $X$ 具有概率密度函数
$$f(x) = \begin{cases}
x, & 0 \le x \le 1, \\
2-x, & 1 < x \le 2, \\
0, & \text{其他},
\end{cases}$$
求 $E(X)$。

**18. 解题思路**：连续型随机变量数学期望公式 $E(X) = \int_{-\infty}^{\infty} x f(x) dx$。
**解答**：
$$E(X) = \int_0^1 x \cdot x dx + \int_1^2 x(2-x) dx = \int_0^1 x^2 dx + \int_1^2 (2x - x^2) dx$$
$$= \left[ \frac{x^3}{3} \right]_0^1 + \left[ x^2 - \frac{x^3}{3} \right]_1^2 = \frac{1}{3} + \left( 4 - \frac{8}{3} \right) - \left( 1 - \frac{1}{3} \right) = \frac{1}{3} + \frac{4}{3} - \frac{2}{3} = 1$$

$$
f(x)=
\begin{cases}
x, &0<x\le1\\
2-x, &1<x\le2\\
0, &\text{其他}
\end{cases}
$$
求$\(E(X)\)$。**思路**：连续期望$\(E(X)\)=\int_{-\infty}^{+\infty}xf(x)dx$分段积分
$$
\begin{align*}
\(E(X)\)&=\int_0^1 x\cdot x dx+\int_1^2 x(2-x)dx\\
&=\int_0^1 x^2 dx+\int_1^2(2x-x^2)dx\\
&=\frac13+\left[x^2-\frac{x^3}{3}\right]_1^2=\frac13+\left(4-\frac83-\left(1-\frac13\right)\right)=\boldsymbol{1}
\end{align*}
$$


## 习题19

已知随机变量 $X$ 的概率密度函数为 $f(x)=\frac{1}{\sqrt{\pi}}e^{-x^2+2x-1}$，求 $E(X), D(X)$。

**19. 解题思路**：将概率密度函数配方，化为正态分布的标准形式，从而直接读出期望和方差。
**解答**：
$$f(x) = \frac{1}{\sqrt{\pi}} e^{-x^2+2x-1} = \frac{1}{\sqrt{\pi}} e^{-(x-1)^2}$$
对比正态分布密度函数 $f(x) = \frac{1}{\sqrt{2\pi}\sigma} e^{-\frac{(x-\mu)^2}{2\sigma^2}}$，可得 $\mu = 1$，$2\sigma^2 = 1 \implies \sigma^2 = 1/2$。
所以 $E(X) = 1$，$D(X) = 1/2$。

$f(x)=\frac1{\sqrt{\pi}}e^{-x^2-2x-1}$，求$\(E(X)\),D(X)$
**思路**：配方法化成正态密度形式。
$$
-x^2-2x-1=-(x^2+2x+1)=-(x+1)^2
$$
$$
f(x)=\frac1{\sqrt{\pi}}e^{-(x+1)^2}=\frac1{\sqrt{2\pi}\cdot\frac1{\sqrt2}}\exp\left(-\frac{(x+1)^2}{2(\frac1{\sqrt2})^2}\right)
$$
$X\sim N(\mu=-1,\sigma^2=\frac12)$
$$
\boldsymbol{\(E(X)\)=-1},\quad \boldsymbol{D(X)=\frac12}
$$


## 习题20

已知连续型随机变量 $X$ 的概率密度函数为
$$f(x) = \begin{cases}
ax+b, & 1 \le x \le 2, \\
0, & \text{其他}
\end{cases}$$
且 $E(X)=\frac{19}{12}$，求 $D(4X+1)$。

**20. 解题思路**：利用密度函数的归一性 $\int f(x) dx = 1$ 和已知期望 $E(X)$ 求出 $a, b$，再根据方差性质求 $D(4X+1) = 16D(X)$。
**解答**：
由 $\int_1^2 (ax+b) dx = 1$ 得：$\frac{3a}{2} + b = 1$ ①
由 $E(X) = \int_1^2 x(ax+b) dx = \frac{19}{12}$ 得：$\frac{7a}{3} + \frac{3b}{2} = \frac{19}{12}$ ②
联立①②解得：$a = 1, b = -1/2$。
$$E(X^2) = \int_1^2 x^2(x - 1/2) dx = \int_1^2 (x^3 - \frac{1}{2}x^2) dx = \left[ \frac{x^4}{4} - \frac{x^3}{6} \right]_1^2 = \frac{31}{12}$$
$$D(X) = E(X^2) - [E(X)]^2 = \frac{31}{12} - \left( \frac{19}{12} \right)^2 = \frac{11}{144}$$
$$D(4X+1) = 16D(X) = 16 \times \frac{11}{144} = \frac{11}{9}$$

$$
f(x)=
\begin{cases}
ax+b, &1\le x\le2\\
0, &\text{其他}
\end{cases}
$$
已知$\(E(X)\)=\frac{19}{12}$，求$\(D(4X+1)\)$
**思路**：归一条件$\int_1^2 (ax+b)dx=1$，$\(E(X)\)=\int_1^2 x(ax+b)dx=\frac{19}{12}$；$D(aX+b)=a^2 D(X)$。
$$
\int_1^2(ax+b)dx=\left[\frac a2 x^2+bx\right]_1^2=\frac{3a}{2}+b=1
$$
$$
\int_1^2 x(ax+b)dx=\left[\frac a3x^3+\frac b2x^2\right]_1^2=\frac{7a}{3}+\frac{3b}{2}=\frac{19}{12}
$$
联立解得$a=1,b=-\frac12$
$$
E(X^2)=\int_1^2 x^2(x-\frac12)dx=\frac{31}{12}
$$
$$
D(X)=E(X^2)-[\(E(X)\)]^2=\frac{31}{12}-\left(\frac{19}{12}\right)^2=\frac{11}{144}
$$
$$
\(D(4X+1)\)=4^2 D(X)=16\cdot\frac{11}{144}=\boldsymbol{\frac{11}{9}}
$$

## 习题22

设随机变量 $X \sim U(0,\pi), Y=\sin X$，求 $E(Y)$。

**22. 解题思路**：$X \sim U(0,\pi)$，概率密度为 $f_X(x) = 1/\pi$。求随机变量函数 $Y=\sin X$ 的期望：$E(Y) = \int_0^{\pi} \sin x \cdot \frac{1}{\pi} dx$。
**解答**：
$$E(Y) = E(\sin X) = \int_0^{\pi} \sin x \cdot \frac{1}{\pi} dx = \frac{1}{\pi} [-\cos x]_0^{\pi} = \frac{1}{\pi} (-(-1) - (-1)) = \frac{2}{\pi}$$


$X\sim U(0,\pi),Y=\sin X$，求$\(E(Y)\)$
**思路**：连续随机变量函数期望$E(g(X))=\int \(g(x)\)f(x)dx$，$f(x)=\frac1\pi,0<x<\pi$
$$
\(E(Y)\)=\int_0^\pi \sin x\cdot\frac1\pi dx=\frac1\pi\left[-\cos x\right]_0^\pi=\frac1\pi(-\cos\pi+\cos0)=\boldsymbol{\frac{2}{\pi}}
$$


## 习题23

某公共汽车站于每时的 10 分，30 分，55 分发车，设乘客不知发车时间，在每时内的任一时刻到达车站，求乘客等候时间的数学期望（精确到秒）。


**23. 解题思路**：设乘客到达时刻 $T \sim U(0,60)$（单位：分钟）。发车时间为10, 30, 55。写出等车时间 $Y$ 关于 $T$ 的分段函数，再求期望。
**解答**：
乘客在 $[0,60)$ 内任意时刻到达是等可能的，概率密度为 $1/60$。
等车时间 $Y$ 与到达时刻 $t$ 的关系：
- 若 $0 \le t < 10$，$Y = 10 - t$
- 若 $10 \le t < 30$，$Y = 30 - t$
- 若 $30 \le t < 55$，$Y = 55 - t$
- 若 $55 \le t < 60$，$Y = 70 - t$
$$E(Y) = \int_0^{10} (10-t)\frac{1}{60} dt + \int_{10}^{30} (30-t)\frac{1}{60} dt + \int_{30}^{55} (55-t)\frac{1}{60} dt + \int_{55}^{60} (70-t)\frac{1}{60} dt$$
分别计算积分：
$\int_0^{10} (10-t) dt = 50$
$\int_{10}^{30} (30-t) dt = 200$
$\int_{30}^{55} (55-t) dt = 312.5$
$\int_{55}^{60} (70-t) dt = 62.5$
总和为 $50 + 200 + 312.5 + 62.5 = 625$。
$$E(Y) = \frac{625}{60} = \frac{125}{12} \text{ 分钟} = 625 \text{ 秒}$$

车站每小时10分、30分、55分发车；乘客任意时刻到达，求等候时间期望。
**思路**：设到达时刻$X\sim U[0,60]$（单位：分钟），分段写出等候时间函数$\(g(x)\)$，再积分求期望。
- $0\le x<10$：等$10-x$
- $10\le x<30$：等$30-x$
- $30\le x<55$：等$55-x$
- $55\le x<60$：等$60-x+10=70-x$
$$
E=\frac1{60}\left[\int_0^{10}(10-x)dx+\int_{10}^{30}(30-x)dx+\int_{30}^{55}(55-x)dx+\int_{55}^{60}(70-x)dx\right]
$$
计算积分：
$$
\begin{align*}
\int_0^{10}(10-x)dx&=50\\
\int_{10}^{30}(30-x)dx&=200\\
\int_{30}^{55}(55-x)dx&=312.5\\
\int_{55}^{60}(70-x)dx&=37.5
\end{align*}
$$
总和$\(=600\)$，$E=\frac{600}{60}=10$分钟。
10分钟 = **600秒**。


## 习题24

设随机变量 $X$ 的概率分布列为
| $X$ | $-1$ | $0$ | $0.5$ | $3$ | $4$ | $6$ |
|---|---|---|---|---|---|---|
| $P\{X=x_i\}$ | $\frac{1}{12}$ | $\frac{1}{12}$ | $\frac{1}{6}$ | $\frac{1}{6}$ | $\frac{1}{3}$ | $\frac{1}{6}$ |

求 $M_e$。

**24. 解题思路**：求离散型随机变量的中位数 $M_e$，即找到使累积概率 $P\{X \le x\} \ge 0.5$ 的最小 $x$ 值。
**解答**：
计算累积概率：
$$P\{X \le -1\} = 1/12 \approx 0.083$$
$$P\{X \le 0\} = 1/12 + 1/12 = 1/6 \approx 0.167$$
$$P\{X \le 0.5\} = 1/6 + 1/6 = 1/3 \approx 0.333$$
$$P\{X \le 3\} = 1/3 + 1/6 = 1/2 = 0.5$$
因为 $P\{X \le 3\} = 0.5$，满足中位数条件，所以中位数 $M_e = 3$。

|---|---|---|---|---|---|---|
|$P$|$\frac1{12}$|$\frac1{12}$|$\frac16$|$\frac16$|$\frac13$|$\frac16$|
求$M_X$（**原点矩$M_X=\(E(X)\)$**，一阶原点矩）
$$
\begin{align*}
\(E(X)\)&=(-1)\cdot\frac1{12}+0\cdot\frac1{12}+0.5\cdot\frac16+3\cdot\frac16+4\cdot\frac13+6\cdot\frac16\\
