### 【例7】
**解题思路**：这是一个放回抽样的模型，取出黑球数 $X$ 服从参数为 $p$ 的几何分布（取值为0,1,2,...）。其中 $p$ 为取得白球的概率。利用几何分布的期望和方差公式（或级数求和）即可求解。
**解答**：
设 $X$ 表示取出的黑球数，每次取得白球的概率为 $p = \frac{m}{m+n}$，取得黑球的概率为 $q = \frac{n}{m+n} = 1-p$。
随机变量 $X$ 的概率函数为：
$$P\{X=k\} = q^k p, \quad k=0,1,2,\cdots$$
**数学期望**：
$$E(X) = \sum_{k=0}^{\infty} k q^k p = pq \sum_{k=1}^{\infty} k q^{k-1} = pq \left( \sum_{k=1}^{\infty} q^k \right)' = pq \left( \frac{q}{1-q} \right)' = pq \frac{1}{(1-q)^2} = \frac{q}{p} = \frac{n}{m}$$
**方差**：
先求二阶原点矩 $E(X^2)$：
$$E(X^2) = \sum_{k=0}^{\infty} k^2 q^k p = \sum_{k=1}^{\infty} (k^2-k) q^k p + \sum_{k=1}^{\infty} k q^k p = q^2 p \sum_{k=2}^{\infty} k(k-1)q^{k-2} + E(X)$$
$$= q^2 p \left( \sum_{k=2}^{\infty} q^k \right)'' + \frac{n}{m} = q^2 p \left( \frac{q^2}{1-q} \right)'' + \frac{n}{m} = q^2 p \frac{2}{(1-q)^3} + \frac{n}{m} = 2 \left( \frac{q}{p} \right)^2 + \frac{n}{m} = \frac{2n^2}{m^2} + \frac{n}{m}$$
从而：
$$D(X) = E(X^2) - [E(X)]^2 = \frac{2n^2}{m^2} + \frac{n}{m} - \frac{n^2}{m^2} = \frac{n^2 + mn}{m^2} = \frac{n(m+n)}{m^2}$$

## 习题17
$$\(P(X=k)\)=\frac{a^k}{(1+a)^{k+1}},\ k=0,1,2\cdots,\ a>0$$求$\(E(X)\)$
**思路**：离散期望定义，令$q=\frac{a}{1+a}$，$\(P(X=k)\)=\frac{1}{1+a}\cdot q^k$；级数$\sum_{k=0}^{\infty}k q^k=\frac{q}{(1-q)^2}$
$$
\(E(X)\)=\sum_{k=0}^{\infty}k\cdot \frac{a^k}{(1+a)^{k+1}}=\frac{1}{1+a}\sum_{k=1}^{\infty}k\left(\frac{a}{1+a}\right)^k
$$
$q=\frac{a}{1+a},1-q=\frac{1}{1+a}$
$$
\(E(X)\)=\frac{1}{1+a}\cdot\frac{q}{(1-q)^2}=\frac{1}{1+a}\cdot \frac{\frac{a}{1+a}}{\left(\frac{1}{1+a}\right)^2}=\boldsymbol{a}
$$

## 习题20
$$
f(x)=
\begin{cases}
ax+b, &1\le x\le2\\
0, &\text{其他}
\end{cases}
$$
已知$\(E(X)\)=\frac{19}{12}$，求$\(D(4X+1)\)$
**思路**：归一条件$\int_1^2 (ax+b)dx=1$，$\(E(X)\)=\int_1^2 x(ax+b)dx=\frac{19}{12}$；$D(aX+b)=a^2 D(X)$。

\int_1^2(ax+b)dx=\left[\frac a2 x^2+bx\right]_1^2=\frac{3a}{2}+b=1
$$

\int_1^2 x(ax+b)dx=\left[\frac a3x^3+\frac b2x^2\right]_1^2=\frac{7a}{3}+\frac{3b}{2}=\frac{19}{12}

## 习题23
车站每小时10分、30分、55分发车；乘客任意时刻到达，求等候时间期望。
**思路**：设到达时刻$X\sim U[0,60]$（单位：分钟），分段写出等候时间函数$\(g(x)\)$，再积分求期望。
- $0\le x<10$：等$10-x$
- $10\le x<30$：等$30-x$
- $30\le x<55$：等$55-x$
- $55\le x<60$：等$60-x+10=70-x$
$$
E=\frac1{60}\left[\int_0^{10}(10-x)dx+\int_{10}^{30}(30-x)dx+\int_{30}^{55}(55-x)dx+\int_{55}^{60}(70-x)dx\right]
$$
计算积分：
$$
\begin{align*}
\int_0^{10}(10-x)dx&=50\\
\int_{10}^{30}(30-x)dx&=200\\
\int_{30}^{55}(55-x)dx&=312.5\\
\int_{55}^{60}(70-x)dx&=37.5
\end{align*}
$$
总和$\(=600\)$，$E=\frac{600}{60}=10$分钟。
10分钟 = **600秒**。
