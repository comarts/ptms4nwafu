### 第一部分：思考题与简答题（第1-8题）

**1. 简述方差分析的基本思路及方差分析对试验数据的基本要求。**
*   **解题思路**：方差分析的核心是将数据的总变异分解为组间变异和组内变异，通过比较两者的均方（F统计量）来判断不同总体均值是否存在显著差异。
*   **解答**：
    *   **基本思路**：将试验数据的总离差平方和（SST）分解为处理效应（SSA）和随机误差（SSE），构造 $F = \frac{MSA}{MSE}$ 统计量进行假设检验。
    *   **基本要求**：试验数据需满足三个基本假定：① 独立性（各观测值相互独立）；② 正态性（各处理下的总体服从正态分布）；③ 方差齐性（各处理下的总体方差相等）。

**2. 多重比较的 $LSR$ 法、$SSR$ 法、$q$ 法的适用条件是什么？**
*   **解答**：
    *   **$q$ 法（Tukey法）**：适用于各样本容量相等的情况，要求严格。
    *   **$SSR$ 法（Duncan法）**：适用于各样本容量相等的情况，本质上是$q$法的修正，检验尺度比$q$法略宽，常用于农业和生物试验。
    *   **$LSR$ 法（最小显著极差法）**：可以用于各样本容量不等的情况。通常包含$q$检验和$SSR$检验两种形式。

**3. 试对双因素方差分析固定模型F分布推导证明。**
*   **解题思路**：基于平方和分解定理和卡方分布的性质。
*   **证明简述**：
    设模型为 $y_{ij} = \mu + \alpha_i + \beta_j + \varepsilon_{ij}$，$\varepsilon_{ij} \sim N(0, \sigma^2)$。总平方和 $SST = SSA + SSB + SSE$。
    在原假设成立下（即无处理效应，$\alpha_i=0, \beta_j=0$），有：
    $$ \frac{SSA}{\sigma^2} \sim \chi^2(a-1), \quad \frac{SSB}{\sigma^2} \sim \chi^2(b-1), \quad \frac{SSE}{\sigma^2} \sim \chi^2((a-1)(b-1)) $$
    且三者相互独立。因此，构造F统计量：
    $$ F_A = \frac{SSA / (a-1)}{SSE / ((a-1)(b-1))} \sim F(a-1, (a-1)(b-1)) $$
    同理可得 $F_B \sim F(b-1, (a-1)(b-1))$。由于期望的比值在无效应时为1附近，若F值过大则拒绝原假设。

**4. 写出一元线性回归模型，并说明回归分析的基本思想。**
*   **解答**：
    *   **模型**：$y = \beta_0 + \beta_1 x + \varepsilon, \quad \varepsilon \sim N(0, \sigma^2)$。
    *   **基本思想**：利用最小二乘法（OLS），寻找一条直线（回归方程），使得所有观测点到该直线的垂直距离（残差）的平方和达到最小，即 $\min \sum (y_i - \hat{y}_i)^2$。

**5. 在一元线性回归方程...中，已知 $\bar{x}=5$，则 $\bar{y}=$ ____。**
*   **思路**：回归方程必定经过样本中心点 $(\bar{x}, \bar{y})$。
*   **解答**：
    将 $\bar{x}=5$ 代入方程 $\hat{y} = 2 + 4x$，得到 $\bar{y} = 2 + 4 \times 5 = 22$。
    **答案：22**

**6. 已知样本数据...计算得到 $l_{xx}=6$，则回归离差平方和 $U=$ ____。**
*   **思路**：题中所述的 $l_{xx}$ 实际上应为 $l_{xy} = \sum x_i y_i - n\bar{x}\bar{y}$。回归离差平方和（回归平方和）公式为 $U = \hat{\beta}_1 l_{xy}$。
*   **解答**：
    已知 $\hat{\beta}_1 = 4$，且 $l_{xy} = 6$。
    $U = \hat{\beta}_1 l_{xy} = 4 \times 6 = 24$。
    **答案：24**

**7. 在单因素 $a$ 水平等重复的方差分析中，计算 $F$ 统计量，其分子与分母的自由度各为（ ）。**
*   **解答**：
    分子为处理间均方 $MSA$，自由度 $df_A = a - 1$。
    分母为误差均方 $MSE$，自由度 $df_E = a(r - 1)$。
    **答案：(C) $a-1, a(r-1)$**

**8. 在一元线性回归模型下，利用样本相关系数进行线性回归显著性检验，随机变量 $Y$ 与普通变量 $x$ 的样本相关系数 $r$ 的表达式为（ ）。**
*   **解答**：
    样本相关系数的定义为 $r = \frac{l_{xy}}{\sqrt{l_{xx} l_{yy}}}$。
    **答案：(D)**

### 第二部分：综合计算题详解（第9-20题）

**9. 不同整地深度对苗高影响是否显著？**
*   **解题思路**：单因素方差分析。因素A（整地深度）有4个水平，每个水平重复5次。
*   **解答**：
    1.  计算各水平均值：$\bar{x}_{15}=70.4$, $\bar{x}_{20}=76.4$, $\bar{x}_{25}=84.8$, $\bar{x}_{30}=97.8$。
    2.  总均值：$\bar{x}=82.35$。
    3.  计算平方和：
        $SSA = 5 \times [(70.4-82.35)^2 + (76.4-82.35)^2 + (84.8-82.35)^2 + (97.8-82.35)^2] \approx 2114.5$
        $SSE = \sum (x_{ij} - \bar{x}_i)^2 = 85.0 + 91.2 + 92.32 + 118.8 = 387.32$
    4.  计算自由度：$df_A = 3$, $df_E = 16$。
    5.  计算均方与F值：$MSA = 2114.5/3 \approx 704.8$，$MSE = 387.32/16 \approx 24.2$。
        $F = 704.8 / 24.2 \approx 29.1$。
    6.  **结论**：查表 $F_{0.05}(3, 16) = 3.24$。由于 $29.1 > 3.24$，$P<0.05$，拒绝原假设。**不同整地深度对苗高有显著影响**。

**10. 黄山松3个家系间苗高差异是否显著？**
*   **思路**：单因素方差分析，但各水平重复数不等。
*   **解答**：
    各组均值：$\bar{x}_A = 25.5$, $\bar{x}_B = 29.6$, $\bar{x}_C = 25.57$。总均值 $\bar{x} \approx 27.04$。
    $SSA = 8 \times (25.5-27.04)^2 + 10 \times (29.6-27.04)^2 + 7 \times (25.57-27.04)^2 \approx 99.63$。
    $SSE \approx 196.4$。自由度 $df_A=2$, $df_E=22$。
    $MSA = 49.82$, $MSE = 8.93$。
    $F = 49.82 / 8.93 \approx 5.58$。
    **结论**：$F_{0.05}(2, 22) = 3.44$。$F > F_{0.05}$，**不同家系间苗高差异显著**。

**11. 4种落叶松苗高生长差异是否显著？若显著，进行多重比较。**
*   **思路**：双因素无重复方差分析（区组-处理）。
*   **解答**：
    计算得：$SSA (处理) = 133.93$, $SSB (区组) = 22.07$, $SSE = 26.77$。
    自由度：$df_A=3$, $df_B=4$, $df_E=12$。
    $MSA = 44.64$, $MSE = 2.23$。$F_A = 44.64/2.23 \approx 20.0$。
    **结论**：$F_{0.05}(3, 12) = 3.49$，$F_A > 3.49$，**4种落叶松苗高生长差异显著**。
    多重比较（LSD法）：$LSD = t_{0.025}(12) \sqrt{2MSE/r} = 2.179 \times \sqrt{2 \times 2.23 / 5} \approx 2.06$。
    均值差：A与D差6.04，B与D差6.4，C与D差5.26，均大于2.06。**A、B、C三种与D存在显著差异，A、B、C间无显著差异**。

**12. 不同土壤、不同肥料及交互作用对苗高是否有显著影响？**
*   **思路**：双因素有重复方差分析。
*   **解答**：
    1.  计算总平方和 $SST = 25347.89$。
    2.  土壤间平方和 $SS_{Soil} \approx 3436.8$（$df=2$）。
    3.  肥料间平方和 $SS_{Fert} \approx 11695.5$（$df=3$）。
    4.  交互作用平方和 $SS_{Int} = SS_{Cell} - SS_{Soil} - SS_{Fert}$，其中 $SS_{Cell} = 13423.8$，得 $SS_{Int} \approx 1708.5$（$df=6$）。

**18. 证明题。**
*   **证明**：

    期望 $E(\hat{\beta}_1) = \sum k_i E(y_i) = \sum k_i (\beta_0 + \beta_1 x_i) = \beta_1 \sum k_i x_i = \beta_1$（因为 $\sum k_i = 0$ 且 $\sum k_i x_i = 1$）。
    方差 $Var(\hat{\beta}_1) = \sum k_i^2 Var(y_i) = \sigma^2 \sum \frac{(x_i - \bar{x})^2}{[\sum (x_i - \bar{x})^2]^2} = \frac{\sigma^2}{\sum (x_i - \bar{x})^2}$。
    故 $\hat{\beta}_1 \sim N\left(\beta_1, \frac{\sigma^2}{\sum (x_i - \bar{x})^2}\right)$。
    (2) $Cov(\hat{\beta}_0, \hat{\beta}_1) = Cov(\bar{y} - \hat{\beta}_1 \bar{x}, \hat{\beta}_1) = Cov(\bar{y}, \hat{\beta}_1) - \bar{x} Var(\hat{\beta}_1)$。
    而 $Cov(\bar{y}, \hat{\beta}_1) = Cov(\frac{1}{n}\sum y_i, \sum k_i y_i) = \frac{\sigma^2}{n} \sum k_i = 0$。
    所以 $Cov(\hat{\beta}_0, \hat{\beta}_1) = 0 - \bar{x} \frac{\sigma^2}{\sum (x_i - \bar{x})^2} = -\frac{\bar{x}\sigma^2}{\sum (x_i - \bar{x})^2}$。证毕。

## 1. 简述方差分析的基本思路及方差分析对试验数据的基本要求
**思路**：
方差分析把**总离差平方和**分解为**因素引起的组间平方和**与**随机误差引起的组内平方和**。构造 $F=\dfrac{S_A/f_A}{S_E/f_E}$，$S_A$ 为因素平方和，$S_E$ 误差平方和；$f_A,f_E$ 是对应自由度。比较 $F$ 和临界值，判断因素各水平均值是否有显著差异。
**基本要求（二条件）**：
1. 独立性：样本相互独立；
\(\boldsymbol{2}\). 正态性：每一水平下观测值服从正态分布；
3. 等方差：各水平总体方差相同。

## \(\boldsymbol{2}\). 多重比较的LSR法、SSR法、q法的适用条件是什么？
- **q法（Tukey法）**：适合**全部两两比较**，各组重复数相等；检验尺度大，比较保守。
- **SSR法（新复极差法）**：适合**全部两两比较**，重复数可等或不等，介于q法与LSR之间。
- **LSR法（最小显著极差法，包含SSR,q）**：一般泛指极差类多重比较；狭义LSR常指SSR；适合多水平均值两两比较。
> 一句话总结：三者都用于单因素方差分析显著后，**多个水平均值两两比较**；q法适合等重复全部两两比较；SSR适用更广。

## 3. 试对双因素方差分析固定模型F分布推导证明
**模型**（无交互固定模型）
$$

$$
$\sum\alpha_i=0,\sum\beta_j=0,\varepsilon_{ij}\sim N(0,\sigma^\(\boldsymbol{2}\))$独立。
平方和分解：
$$

$$
$$
\begin{aligned}

S_E&=\sum_{i,j}(x_{ij}-\bar x_{i\cdot}-\bar x_{\cdot j}+\bar x_{\cdot\cdot})^\(\boldsymbol{2}\),\quad f_E=(a-1)(b-1)
\end{aligned}
$$
由卡方分布性质：
- $H_{0A}:\alpha_1=\dots=\alpha_a=0$成立时，$\dfrac{S_A}{\sigma^\(\boldsymbol{2}\)}\sim\chi^\(\boldsymbol{2}\)(a-1)$
- $H_{0B}:\beta_1=\dots=\beta_b=0$成立时，$\dfrac{S_B}{\sigma^\(\boldsymbol{2}\)}\sim\chi^\(\boldsymbol{2}\)(b-1)$

$$
F_A=\frac{S_A/f_A}{S_E/f_E}\sim F(a-1,(a-1)(b-1))
$$
$$
F_B=\frac{S_B/f_B}{S_E/f_E}\sim F(b-1,(a-1)(b-1))
$$
> 含交互模型同理，自由度$f_E=(a-1)(b-1)$，交互项$S_{A\times B},f=(a-1)(b-1)$。

##18.证明题：一元线性回归最小二乘估计
模型：$y_i=\beta_0+\beta_1 x_i+\varepsilon_i,\varepsilon_i\sim N(0,\sigma^\(\boldsymbol{2}\))$独立
$$
\hat\beta_1=\frac{\sum(x_i-\bar x)(y_i-\bar y)}{\sum(x_i-\bar x)^\(\boldsymbol{2}\)}=\sum w_i y_i,\ w_i=\frac{x_i-\bar x}{\sum(x_i-\bar x)^\(\boldsymbol{2}\)}
$$
(1) 证明$\hat\beta_1\sim N\left(\beta_1,\frac{\sigma^\(\boldsymbol{2}\)}{\sum_{i=1}^n(x_i-\bar x)^\(\boldsymbol{2}\)}\right)$
> $\hat\beta_1$是$y_i$线性组合，正态变量线性组合仍然正态。
$$
E(\hat\beta_1)=\beta_1,\quad D(\hat\beta_1)=\sum w_i^\(\boldsymbol{2}\) D(y_i)=\sigma^\(\boldsymbol{2}\)\sum w_i^\(\boldsymbol{2}\)=\frac{\sigma^\(\boldsymbol{2}\)}{\sum(x_i-\bar x)^\(\boldsymbol{2}\)}
$$
(\(\boldsymbol{2}\))$\mathrm{Cov}(\hat\beta_0,\hat\beta_1)=-\frac{\bar x}{\sum(x_i-\bar x)^\(\boldsymbol{2}\)}\sigma^\(\boldsymbol{2}\)$
$\hat\beta_0=\bar y-\hat\beta_1\bar x$
$$
\mathrm{Cov}(\hat\beta_0,\hat\beta_1)=\mathrm{Cov}(\bar y-\bar x\hat\beta_1,\hat\beta_1)=\mathrm{Cov}(\bar y,\hat\beta_1)-\bar x D(\hat\beta_1)
$$
$\mathrm{Cov}(\bar y,\hat\beta_1)=0$，于是
$$
\mathrm{Cov}(\hat\beta_0,\hat\beta_1)=-\bar x\cdot\frac{\sigma^\(\boldsymbol{2}\)}{\sum(x_i-\bar x)^\(\boldsymbol{2}\)}
$$
