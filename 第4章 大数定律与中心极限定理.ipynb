## 习题1
设随机变量 $X$ 的数学期望 $E(X)=5$，方差 $D(X)=2$，利用切比雪夫不等式估计下列概率：

(1) $P\{3<X<7\}$； (2) $P\{-7<X<17\}$； (3) $P\{|X-5|>8\}$。

解题思路： 利用切比雪夫不等式 $P\{|X - E(X)| \ge \varepsilon\} \le \dfrac{D(X)}{\varepsilon^2}$ 或其等价形式 $P\{|X - E(X)| < \varepsilon\} \ge 1 - \frac{D(X)}{\varepsilon^2}$。

解答：

(1) $P\{3 < X < 7\} = P\{|X - 5| < 2\} \ge 1 - \dfrac{D(X)}{2^2} = 1 - \dfrac{2}{4} = 0.5$

(2) $P\{-7 < X < 17\} = P\{|X - 5| < 12\} \ge 1 - \dfrac{D(X)}{12^2} = 1 - \dfrac{2}{144} = \dfrac{71}{72} \approx 0.9861$

(3) $P\{|X - 5| > 8\} \le \dfrac{D(X)}{8^2} = \dfrac{2}{64} = \dfrac{1}{32} = 0.03125$

## 习题2

设随机变量 $X$ 在区间 $[0,10]$ 上服从均匀分布，求 $E(X)$，$D(X)$，并利用切比雪夫不等式估计 $P\{|X-E(X)|>4\}$，再计算出 $P\{|X-E(X)|>4\}$ 的精确值。

解题思路：均匀分布 $X \sim U(a,b)$ 的期望 $E(X) = \frac{a+b}{2}$，方差 $D(X) = \frac{(b-a)^2}{12}$。先用切比雪夫不等式估计上限，再用概率密度函数积分求精确值。

解答：

由 $X \sim U(0,10)$ 可得：
$E(X) = \frac{0+10}{2} = 5$，$D(X) = \frac{(10-0)^2}{12} = \frac{100}{12} = \frac{25}{3}$

切比雪夫不等式估计：

$P\{|X - 5| > 4\} \le \frac{D(X)}{4^2} = \frac{25/3}{16} = \frac{25}{48} \approx 0.5208$

精确值计算：

$P\{|X - 5| > 4\} = P\{X < 1\} + P\{X > 9\} = \int_0^1 \frac{1}{10} dx + \int_9^{10} \frac{1}{10} dx = 0.1 + 0.1 = 0.2$

答案：$E(X)=5,D(X)=\dfrac{25}{3}$；切比雪夫估计 $P\le \dfrac{25}{48}$；精确值 $\dfrac15$


## 习题3

设随机变量 $X \sim U(-1,b)$，若由切比雪夫不等式有 $P\{|X-1|<\varepsilon\} \ge \dfrac{2}{3}$，则 $b=\_\_\_$，$\varepsilon=\_\_\_$。

解题思路：由 $X \sim U(-1,b)$ 可得期望 $E(X) = \dfrac{b-1}{2}$。观察不等式 $|X - 1| < \varepsilon$ 可知期望为 $1$，从而求出 $b$。利用 $P\{|X - E(X)| < \varepsilon\} \ge 1 - \dfrac{D(X)}{\varepsilon^2}$ 求解 $\varepsilon$。

解答：

$E(X) = \dfrac{-1+b}{2} = 1 \Rightarrow b = 3$

$D(X) = \frac{(3 - (-1))^2}{12} = \dfrac{16}{12} = \dfrac{4}{3}$

根据切比雪夫不等式：$P\{|X - 1| < \varepsilon\} \ge 1 - \dfrac{4/3}{\varepsilon^2} = \dfrac{2}{3}$

解得 $\dfrac{4/3}{\varepsilon^2} = \dfrac{1}{3} \Rightarrow \varepsilon^2 = 4 \Rightarrow \varepsilon = 2$（因 $\varepsilon>0$）

填空： $b=3$，$\varepsilon=2$

## 习题4

设 $\{X_n\}$ 是相互独立同分布的随机变量序列，其分布函数为 $F(x)=\dfrac{1}{2}+\dfrac{1}{\pi}\arctan x$，$-\infty<x<+\infty$，试问 $\{X_n\}$ 是否服从大数定律。

解题思路： 判断辛钦大数定律的条件，即独立同分布且期望存在。计算该分布的期望，若期望不存在（发散），则不服从大数定律。这其实是柯西分布。

解答：

该分布的概率密度函数为 $f(x) = F'(x) = \dfrac{1}{\pi(1+x^2)}$

计算期望：
$$E(X) = \int_{-\infty}^{+\infty} x f(x) dx = \int_{-\infty}^{+\infty} \dfrac{x}{\pi(1+x^2)} dx$$

由于被积函数是奇函数，且 $$\int_0^{+\infty} \dfrac{x}{1+x^2} dx = \dfrac{1}{2} \ln(1+x^2)|_0^{+\infty} = +\infty$$ 发散，故期望 $E(X)$ 不存在

因为辛钦大数定律要求随机变量序列的数学期望存在，所以该序列不服从大数定律。

## 习题5

证明泊松大数定律：设在第 $i$ 次试验中事件 $A$ 发生的概率为 $p_i$，$M_n$ 是 $n$ 次独立试验中事件 $A$ 发生的次数，则对于任意正数 $\varepsilon$，有 $\lim\limits_{n\to\infty} P\left\{\left|\dfrac{M_n}{n}-\dfrac{p_1+p_2+\dots+p_n}{n}\right|<\varepsilon\right\}=1$。

解题思路： 利用切比雪夫不等式。构造独立随机变量 $X_i$ 表示第 $i$ 次试验中 $A$ 发生的次数，$X_i \sim B(1, p_i)$，求出 $M_n/n$ 的期望和方差，代入不等式并取极限。

解答：
设 $X_i = \begin{cases} 1, & \text{第 } i \text{ 次试验 } A \text{ 发生} \\ 0, & \text{第 } i \text{ 次试验 } A \text{ 不发生} \end{cases}$，则 $M_n = \sum\limits_{i=1}^n X_i$

因为 $X_i$ 相互独立，$E(X_i) = p_i$，$D(X_i) = p_i(1-p_i) \le \dfrac{1}{4}$

$$E\left(\dfrac{M_n}{n}\right) = \dfrac{1}{n} \sum_{i=1}^n p_i$$

$$D\left(\frac{M_n}{n}\right) = \frac{1}{n^2} \sum_{i=1}^n p_i(1-p_i) \le \frac{1}{n^2} \cdot n \cdot \frac{1}{4} = \frac{1}{4n}$$

根据切比雪夫不等式：

$$P\left\{\left|\frac{M_n}{n} - \frac{1}{n}\sum_{i=1}^n p_i\right| < \varepsilon\right\} \ge 1 - \frac{D(M_n/n)}{\varepsilon^2} \ge 1 - \frac{1}{4n\varepsilon^2}$$

当 $n \to \infty$ 时，$1 - \dfrac{1}{4n\varepsilon^2} \to 1$，故得证。


## 习题6

若 $X_1, X_2, \dots, X_{1000}$ 为独立同分布的随机变量，$X_i \sim B(1,p)$，$i=1,2,\dots,1000$，则下列选项不正确的是（ ）。

(A) $\dfrac{1}{1000}\sum\limits_{i=1}^{1000} X_i \approx p$

(B) $\sum\limits_{i=1}^{1000} X_i \sim B(1000,p)$

(C) $P\{a<\sum\limits_{i=1}^{1000} X_i<b\} \approx \Phi\left(\dfrac{b-1000p}{\sqrt{1000p(1-p)}}\right)-\Phi\left(\dfrac{a-1000p}{\sqrt{1000p(1-p)}}\right)$

(D) $P\{a<\sum\limits_{i=1}^{1000} X_i<b\} \approx \Phi(b)-\Phi(a)$

解题思路：独立0-1变量之和为二项分布；大数定律；中心极限定理。

解答：

(A) $\dfrac1{1000}\sum X_i\approx p$：大数定律 ✔

(B) $\sum X_i\sim B(1000,p)$：独立0-1和 ✔

(C) 中心极限定理标准化形式 ✔

(D) $P(a<\sum X_i<b)\approx\Phi(b)-\Phi(a)$：错误，没有标准化。

答案：$D$


## 习题7

某汽车销售点每天售出的汽车数服从参数为 $\lambda=2$ 的泊松分布，若该销售点一年365天都经营汽车销售，且每天售出的汽车数是相互独立的，求一年中售出700辆以上汽车的概率。


解题思路：设第 $i$ 天售出量为 $X_i \sim P(2)$，总销量 $S = \sum\limits_{i=1}^{365} X_i$。$E(X_i)=2, D(X_i)=2$。使用中心极限定理，$S$ 近似服从 $N(365 \times 2, 365 \times 2) = N(730, 730)$。

解答：

设 $X_i$ 为第 $i$ 天的售出量，则 $X_i \sim P(2)$，且相互独立。

$E(X_i)=2$，$D(X_i)=2$

总销量 $S = \sum_{i=1}^{365} X_i$，则 $E(S) = 730$，$D(S) = 730$

由中心极限定理，$S$ 近似服从正态分布 $N(730, 730)$

$$P(S > 700) = 1 - P(S \le 700) \approx 1 - \Phi\left(\frac{700 - 730}{\sqrt{730}}\right) = 1 - \Phi\left(\frac{-30}{\sqrt{730}}\right) = \Phi(1.11) = 0.8665$$

注：若使用连续性修正，$P(S > 700) = P(S \ge 701) \approx 1 - \Phi\left(\dfrac{700.5 - 730}{\sqrt{730}}\right) = \Phi(1.09) = 0.8621$。一般教材若不特别说明，可不修正，取 $0.8665$。

## 习题8

设随机变量序列 $\{X_n\}(n \ge 1)$ 相互独立且都服从在 $[-1,1]$ 上的均匀分布，求 $\lim\limits_{n\to\infty} P\left\{\left|\dfrac{\sum\limits_{i=1}^n X_i}{\sqrt{n}}\right| \le 1\right\}$（用 $\Phi(x)$ 表示）。


解题思路： $X_i \sim U(-1,1)$，$E(X_i)=0$，$D(X_i)=\frac{1}{3}$。由中心极限定理，$\frac{\sum_{i=1}^n X_i}{\sqrt{n/3}}$ 近似服从 $N(0,1)$。将所求式子标准化。

解答：

令 $S_n=\sum_{i=1}^n X_i$，则
$$
E(S_n)=0,\qquad \operatorname{Var}(S_n)=n\cdot \frac13=\frac n3.
$$

考虑随机变量 $\dfrac{S_n}{\sqrt n}$，其期望为 $0$，方差为
$$
\operatorname{Var}\left(\frac{S_n}{\sqrt n}\right)=\frac{1}{n}\cdot \frac n3=\frac13.
$$

由中心极限定理，
$$
\frac{S_n/\sqrt n}{\sqrt{1/3}}=\sqrt{3}\,\frac{S_n}{\sqrt n}
$$
依分布收敛于标准正态分布 $N(0,1)$。

所求概率为
$$
P\left\{\left|\frac{S_n}{\sqrt n}\right|\le 1\right\}
= P\left\{-1\le \frac{S_n}{\sqrt n}\le 1\right\}.
$$
将不等式标准化：
$$
-1\le \frac{S_n}{\sqrt n}\le 1
\quad\Longleftrightarrow\quad
-\sqrt3\le \sqrt{3}\,\frac{S_n}{\sqrt n}\le \sqrt3.
$$
因此
$$
\lim_{n\to\infty}P\left\{\left|\frac{S_n}{\sqrt n}\right|\le 1\right\}
= P\{-\sqrt3\le Z\le \sqrt3\},
$$
其中 $Z\sim N(0,1)$。于是
$$
P\{-\sqrt3\le Z\le \sqrt3\}
= \Phi(\sqrt3)-\Phi(-\sqrt3)
= \Phi(\sqrt3)-[1-\Phi(\sqrt3)]
= 2\Phi(\sqrt3)-1.
$$

## 习题9

设随机变量序列 $\{X_n\}(n \ge 1)$ 相互独立且都服从参数为 $\frac{1}{2}$ 的两点分布，如果常数 $C$ 使得 $\lim_{n\to\infty} P\left\{C\frac{\sum_{i=1}^n (X_{2i}-X_{2i-1})}{\sqrt{n}} \le x\right\}=\Phi(x)$，求 $C$。

解题思路： $X_i$ 服从参数为 $1/2$ 的两点分布（即 $P(X_i=1)=1/2, P(X_i=0)=1/2$），则 $E(X_i)=1/2, D(X_i)=1/4$。令 $Y_i = X_{2i} - X_{2i-1}$，求出 $E(Y_i)$ 和 $D(Y_i)$，然后对 $\sum Y_i$ 使用中心极限定理。

解答：
$E(X_i) = \frac{1}{2}$，$D(X_i) = \frac{1}{2} \times \frac{1}{2} = \frac{1}{4}$

令 $Y_i = X_{2i} - X_{2i-1}$，由于各 $X_i$ 独立，故 $Y_i$ 相互独立

$$E(Y_i) = E(X_{2i}) - E(X_{2i-1}) = \frac{1}{2} - \frac{1}{2} = 0$$
$$D(Y_i) = D(X_{2i}) + D(X_{2i-1}) = \frac{1}{4} + \frac{1}{4} = \frac{1}{2}$$
由中心极限定理，$$\frac{\sum\limits_{i=1}^n Y_i - 0}{\sqrt{n \cdot \frac{1}{2}}} = \sqrt{2} \frac{\sum\limits_{i=1}^n Y_i}{\sqrt{n}}$$
近似服从 $N(0,1)$

要使 $C \dfrac{\sum_{i=1}^n Y_i}{\sqrt{n}}$ 近似服从 $N(0,1)$，则 $C = \sqrt{2}$

## 习题10

根据往年的统计资料，某高校大二学生参加全国大学英语四级考试的通过率为 $60\%$，2019年春季该校参加全国大学英语四级考试的大二学生人数为3000，假定每个学生能否通过全国大学英语四级考试相互独立。求2019年春季该校参加全国大学英语四级考试的大二学生中，通过考试人数超过1800的概率。

解题思路： 设通过人数为 $X$，则 $X \sim B(3000, 0.6)$。$E(X)=1800$，$D(X)=3000 \times 0.6 \times 0.4 = 720$。利用棣莫弗-拉普拉斯中心极限定理求 $P(X > 1800)$。

解答：
设通过人数为 $X$，则 $X \sim B(3000, 0.6)$

$E(X) = 3000 \times 0.6 = 1800$，$D(X) = 3000 \times 0.6 \times 0.4 = 720$

由中心极限定理：
$$P(X > 1800) = 1 - P(X \le 1800) \approx 1 - \Phi\left(\frac{1800 - 1800}{\sqrt{720}}\right) = 1 - \Phi(0) = 1 - 0.5 = 0.5$$

若用连续性修正：$P(X > 1800) = P(X \ge 1801) \approx 1 - \Phi\left(\dfrac{1800.5 - 1800}{\sqrt{720}}\right) = 1 - \Phi(0.0186) \approx 0.4925$。

## 习题11

某保险公司有 10000 人参加 60 岁以下人身保险，每人每年付 12 元保险费，若在一年内死亡，受益人可在保险公司领取 1000 元赔偿费。设一个人在一年内死亡的概率为 0.006，求保险公司此项业务一年的利润不少于 50000 元的概率，利润不少于 40000 元的概率又是多少？

解题思路： 设一年内死亡人数 $X \sim B(10000, 0.006)$。计算期望和方差。总收入为 $10000 \times 12 = 120000$ 元。利润 $L = 120000 - 1000X$。将利润不等式转化为死亡人数 $X$ 的不等式，然后用中心极限定理求解。

解答：

设一年内死亡人数为 $X$，则 $X \sim B(10000, 0.006)$

$E(X) = 10000 \times 0.006 = 60$，$D(X) = 10000 \times 0.006 \times 0.994 = 59.64$。$\sqrt{D(X)} \approx 7.7227$

利润 $L = 120000 - 1000X$

(1) 利润不少于 $50000$ 元：$120000 - 1000X \ge 50000 \Rightarrow X \le 70$

$P(X \le 70) \approx \Phi\left(\dfrac{70 - 60}{\sqrt{59.64}}\right) = \Phi(1.2949) \approx 0.9022$

连续性修正：$\Phi(\dfrac{70.5 - 60}{7.7227}) = \Phi(1.3596) \approx 0.9130$

(2) 利润不少于 $40000$ 元：$120000 - 1000X \ge 40000 \Rightarrow X \le 80$

$P(X \le 80) \approx \Phi\left(\dfrac{80 - 60}{\sqrt{59.64}}\right) = \Phi(2.5898) \approx 0.9952$

连续性修正：$\Phi(\frac{80.5 - 60}{7.7227}) = \Phi(2.6545) \approx 0.9960$

## 习题12

历史上，皮尔逊为验证概率与频率的关系，曾掷硬币12000次，正面出现了6019次，如果我们现在重复他的试验，求正面出现的频率与其概率之差的绝对值，不大于当年皮尔逊试验所发生的偏差的概率。

解题思路： 设正面次数 $X \sim B(12000, 0.5)$。当年皮尔逊试验的偏差 $\varepsilon = |\dfrac{6019}{12000} - 0.5| = \dfrac{19}{12000}$。求 $P(|\dfrac{X}{12000} - 0.5| \le \dfrac{19}{12000}) = P(|X - 6000| \le 19)$。使用中心极限定理。

解答：

设正面出现次数为 $X$，则 $X \sim B(12000, 0.5)$

$E(X) = 6000$，$D(X) = 3000$，$\sqrt{D(X)} \approx 54.772$

当年皮尔逊试验的偏差为 $\left|\dfrac{6019}{12000} - 0.5\right| = \dfrac{19}{12000}$

所求概率为 $P\left\{\left|\dfrac{X}{12000} - 0.5\right| \le \dfrac{19}{12000}\right\} = P\{|X - 6000| \le 19\}$

由中心极限定理：
$$P\{|X - 6000| \le 19\} \approx 2\Phi\left(\frac{19}{\sqrt{3000}}\right) - 1 = 2\Phi(0.3469) - 1 \approx 2 \times 0.6357 - 1 = 0.2714$$

连续性修正：$P\{-19.5 \le X - 6000 \le 19.5\} \approx 2\Phi(0.356) - 1 = 0.278$

## 习题13

自由度 $n = 40$

## 习题14

设 $P(A)=p$（$p$未知），做相同试验100次，用 $A$ 发生的频率代替概率 $p$，估计所产生的误差小于 $10\%$ 的概率。（提示：使用 $pq \le \frac{1}{4}$。）

解题思路：设100次试验中 $A$ 发生次数 $X \sim B(100,p)$，频率 $f = X/100$。误差 $|f - p| < 0.1$。利用切比雪夫不等式估计概率下限。

解答：

设 $X$ 为100次试验中 $A$ 发生的次数，$X \sim B(100,p)$

$E(X/100) = p$，$D(X/100) = \frac{p(1-p)}{100}$

由切比雪夫不等式：

$$P\left\{\left|\frac{X}{100} - p\right| < 0.1\right\} \ge 1 - \frac{p(1-p)/100}{0.1^2} = 1 - p(1-p)$$
因为 $p(1-p) \le \frac{1}{4}$，所以 $1 - p(1-p) \ge 1 - \frac{1}{4} = 0.75$

即产生的误差小于 $10\%$ 的概率至少为 0.75。



## 习题15

把一枚均匀对称的硬币独立地投掷12000次，$X$ 表示正面出现的次数，分别用切比雪夫不等式、中心极限定理求解下面三个问题：

(1) 计算 $P\{5800 \le X \le 6200\}$；

(2) 求满足 $P\left\{\left|\frac{X}{12000}-\frac{1}{2}\right|<\Delta\right\} \ge 0.99$ 的 $\Delta$；

(3) 求满足 $P\left\{\left|\frac{X}{n}-\frac{1}{2}\right|<0.005\right\} \ge 0.99$ 的最小值 $n$。


解题思路： $X \sim B(12000, 0.5)$，$E(X)=6000$，$D(X)=3000$。分别代入切比雪夫不等式（利用 $D(X)/\varepsilon^2$）和中心极限定理（标准化后查正态分布表）求解。

解答：

(1) 求 $P\{5800 \le X \le 6200\}$
改写：
$$
P\{5800\le X\le 6200\}=P\{|X-6000|\le 200\}
$$

 ① 切比雪夫不等式

切比雪夫：$P\{|X-E(X)|\ge \varepsilon\}\le \dfrac{D(X)}{\varepsilon^2}$
$\varepsilon=200$
$$
P\{|X-6000|\le 200\}\ge 1-\frac{D(X)}{\varepsilon^2}=1-\frac{3000}{200^2}=1-\frac{3000}{40000}=1-0.075=\boldsymbol{0.925}
$$

② 中心极限定理（棣莫弗–拉普拉斯）
$$
\frac{X-np}{\sqrt{np(1-p)}}=\frac{X-6000}{\sqrt{3000}}\stackrel{d}{\to}N(0,1)
$$
$$
\begin{aligned}
P\{|X-6000|\le 200\}
&=P\left\{\frac{|X-6000|}{\sqrt{3000}}\le \frac{200}{\sqrt{3000}}\right\}\\
&=P\left\{|Z|\le \frac{200}{10\sqrt{30}}\right\}
=P\left\{|Z|\le \frac{20}{\sqrt{30}}\right\}\\
&\approx P\{|Z|\le 3.6515\}\\
&=\Phi(3.6515)-\Phi(-3.6515)=2\Phi(3.6515)-1 \approx 0.9997
\end{aligned}
$$


(2) 求满足 $\displaystyle P\left\{\left|\frac{X}{12000}-\frac12\right|<\Delta\right\}\ge 0.99$ 的 $\Delta$

不等式变形：
$$
\left|\frac{X}{12000}-\frac12\right|<\Delta
\iff |X-6000|<12000\Delta
$$
令 $\varepsilon=12000\Delta$

① 切比雪夫不等式
$$
P\left\{\left|\frac{X}{n}-\frac12\right|<\Delta\right\}\ge 1-\frac{D(\bar X)}{\Delta^2}=0.99
$$
$$
D(\bar X)=\frac{1}{4n}=\frac{1}{4\times 12000}=\frac{1}{48000}
$$
$$

=P\left\{|Z|<\frac{\Delta \cdot n}{\sqrt{np(1-p)}}\right\}\ge 0.99

## 习题16

设随机变量 $X_1, X_2, \dots, X_n$ 相互独立，且均服从正态分布 $N(0,1)$，试证明
$$P\left\{0<\sum_{i=1}^n X_i^2<2n\right\} \ge \frac{n-2}{n}.$$

解题思路：由 $X_i \sim N(0,1)$ 可知 $X_i^2 \sim \chi^2(1)$。由卡方分布可加性，$Y = \sum X_i^2 \sim \chi^2(n)$。求出 $Y$ 的期望和方差，利用切比雪夫不等式证明。

解答：

因为 $X_i \sim N(0,1)$，所以 $X_i^2 \sim \chi^2(1)$

由 $\chi^2$ 分布的可加性，$Y = \sum_{i=1}^n X_i^2 \sim \chi^2(n)$

所以 $E(Y) = n$，$D(Y) = 2n$

利用切比雪夫不等式，对于 $Y$：
$$P\{0 < Y < 2n\} = P\{-n < Y - n < n\} = P\{|Y - n| < n\} \ge 1 - \frac{D(Y)}{n^2} = 1 - \frac{2n}{n^2} = 1 - \frac{2}{n} = \frac{n-2}{n}$$

得证。
