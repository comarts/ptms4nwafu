## 习题1

一个袋内装有5个白球，3个红球，第一次从袋内任意取一个球，不放回，第二次又从袋内任意取两个球，$X_i$表示第$i$次取到的白球数，$i=1,2$。求：(1) $(X_1,X_2)$的概率分布；(2) $P\{X_1=0, X_2 \neq 0\}$，$P\{X_1=X_2\}$，$P\{X_1X_2=0\}$。

解题思路：利用超几何分布求联合概率。第一次取1个球，第二次取2个球，总样本空间。条件概率公式 $P(A|B) = P(AB)/P(B)$。

解答：

(1) 求联合概率 $P(X_1=x_1,X_2=x_2)$

① $P(X_1=0,X_2=0)$
$X_1=0$：第一次取红，剩余：5白，2红（共7）；第二次取2红
$$
P(X_1=0)=\frac{3}{8},\quad P(X_2=0|X_1=0)=\frac{\mathrm{C}_2^2}{\mathrm{C}_7^2}=\frac{1}{21}
$$
$$
P(X_1=0,X_2=0)=\frac{3}{8}\cdot \frac{1}{21}=\frac{3}{168}=\frac{1}{56}
$$

② $P(X_1=0,X_2=1)$
第一次红，剩余5白2红，第二次1白1红：
$$
P(X_2=1|X_1=0)=\frac{\mathrm{C}_5^1\mathrm{C}_2^1}{\mathrm{C}_7^2}=\frac{10}{21}
$$
$$
P(X_1=0,X_2=1)=\frac{3}{8}\cdot\frac{10}{21}=\frac{30}{168}=\frac{5}{28}
$$

③ $P(X_1=0,X_2=2)$
第一次红，剩余5白2红，第二次2白：
$$
P(X_2=2|X_1=0)=\frac{\mathrm{C}_5^2}{\mathrm{C}_7^2}=\frac{10}{21}
$$
$$
P(X_1=0,X_2=2)=\frac{3}{8}\cdot\frac{10}{21}=\frac{30}{168}=\frac{5}{28}
$$

④ $P(X_1=1,X_2=0)$
第一次白，剩余：4白，3红（共7）；第二次2红：
$$
P(X_1=1)=\frac{5}{8},\quad P(X_2=0|X_1=1)=\frac{\mathrm{C}_3^2}{\mathrm{C}_7^2}=\frac{3}{21}=\frac{1}{7}
$$
$$
P(X_1=1,X_2=0)=\frac{5}{8}\cdot\frac{3}{21}=\frac{15}{168}=\frac{5}{56}
$$

⑤ $P(X_1=1,X_2=1)$
第一次白，剩余4白3红，第二次1白1红：
$$
P(X_2=1|X_1=1)=\frac{\mathrm{C}_4^1\mathrm{C}_3^1}{\mathrm{C}_7^2}=\frac{12}{21}=\frac{4}{7}
$$
$$
P(X_1=1,X_2=1)=\frac{5}{8}\cdot\frac{12}{21}=\frac{60}{168}=\frac{5}{14}
$$

⑥ $P(X_1=1,X_2=2)$
第一次白，剩余4白3红，第二次2白：
$$
P(X_2=2|X_1=1)=\frac{\mathrm{C}_4^2}{\mathrm{C}_7^2}=\frac{6}{21}=\frac{2}{7}
$$
$$
P(X_1=1,X_2=2)=\frac{5}{8}\cdot\frac{6}{21}=\frac{30}{168}=\frac{5}{28}
$$

✅ 校验总和：
$$
\frac{1}{56}+\frac{5}{28}+\frac{5}{28}+\frac{5}{56}+\frac{5}{14}+\frac{5}{28}=1
$$

联合分布表：

| $X_1\setminus X_2$ | 0 | 1 | 2 |
|---|---|---|---|
| 0 | $\dfrac{1}{56}$ | $\dfrac{5}{28}$ | $\dfrac{5}{28}$ |
| 1 | $\dfrac{5}{56}$ | $\dfrac{5}{14}$ | $\dfrac{5}{28}$ |

---

(2) 计算三个概率

① $P\{X_1=0,X_2\neq 0\}$
$X_1=0,X_2\neq0$ 即 $(0,1),(0,2)$
$$
P=\frac{5}{28}+\frac{5}{28}=\frac{10}{28}=\frac{5}{14}$$

② $P\{X_1=X_2\}$

满足 $X_1=X_2$ 的点：$(0,0),\;(1,1)$
$$
P=\frac{1}{56}+\frac{5}{14}=\frac{1}{56}+\frac{20}{56}=\frac{21}{56}=\frac{3}{8}$$

③ $P\{X_1X_2=0\}$

$X_1X_2=0 \iff X_1=0$ 或 $X_2=0$
对应点：$(0,0),(0,1),(0,2),(1,0)$
$$
P=\frac{1}{56}+\frac{5}{28}+\frac{5}{28}+\frac{5}{56}=\frac{1+10+10+5}{56}=\frac{26}{56}=\frac{13}{28}$$

## 习题2

设随机向量$(X,Y)$的概率密度函数为
$$f(x,y) = \begin{cases} c, & 0 \le x \le 1, 0 \le y \le 2, \\ 0, & \text{其他}, \end{cases}$$
求 $X$ 与 $Y$ 中至少有一个小于 $\frac{1}{2}$ 的概率。

解题思路：根据联合概率密度函数的归一性求 $c$

计算对立事件的概率：$P(X < \frac{1}{2} \text{ 或 } Y < \frac{1}{2}) = 1 - P(X \ge \frac{1}{2}, Y \ge \frac{1}{2})$


解答：

由 $$\int_{0}^{1} \int_{0}^{2} c \, dy \, dx = 2c = 1 \implies c = \frac{1}{2}$$

$$P(X \ge \frac{1}{2}, Y \ge \frac{1}{2}) = \int_{1/2}^{1} \int_{1/2}^{2} \dfrac{1}{2} \, dy \, dx = \dfrac{1}{2} \times \dfrac{1}{2} \times \dfrac{3}{2} = \dfrac{3}{8}$$

所以所求概率为 $1 - \dfrac{3}{8} = \dfrac{5}{8}$

## 习题3
设二维随机向量 $(X,Y)$ 的概率密度函数为
$$f(x,y) = \frac{1}{2\pi}e^{-\frac{x^2+y^2}{2}}, \quad -\infty < x,y < +\infty,$$
试求二维随机向量 $(X,Y)$ 关于 $X,Y$ 的边缘概率密度函数。

解题思路：二维标准正态分布的边缘分布仍是标准正态分布。利用积分求边缘概率密度函数。

解答：
$$f_X(x) = \int_{-\infty}^{\infty} \frac{1}{2\pi} e^{-\frac{x^2+y^2}{2}} \, dy = \frac{1}{\sqrt{2\pi}} e^{-\frac{x^2}{2}} \int_{-\infty}^{\infty} \frac{1}{\sqrt{2\pi}} e^{-\frac{y^2}{2}} \, dy = \frac{1}{\sqrt{2\pi}} e^{-\frac{x^2}{2}}$$

由对称性，$f_Y(y) = \frac{1}{\sqrt{2\pi}} e^{-\frac{y^2}{2}}$

即 $X\sim N(0,1), Y\sim N(0,1)$


## 习题4

设随机向量$(X,Y)$服从区域$G: \{(x,y) \mid x>y, x^2 < y\}$上的均匀分布，求$(X,Y)$的概率密度函数与边缘概率密度函数。

解题思路：均匀分布的概率密度函数为区域面积的倒数。需先计算区域 $G$ 的面积。

解答：
区域
$$
G=\{(x,y)\mid x>y,\ x^2<y\}
$$
由 $x^2<y<x$ 可得
$$
0<x<1,\qquad x^2<y<x
$$
区域 $G$ 的面积为
$$
S_G=\int_0^1 (x-x^2)\,dx=\frac12-\frac13=\frac16
$$
因为 $(X,Y)$ 在 $G$ 上服从均匀分布，所以联合概率密度为
$$
f_{X,Y}(x,y)=
\begin{cases}
6, & 0<x<1,\ x^2<y<x,\\
0, & \text{其他}
\end{cases}
$$

求 $X$ 的边缘概率密度：
$$
f_X(x)=\int_{-\infty}^{+\infty}f_{X,Y}(x,y)\,dy
$$
当 $0<x<1$ 时，
$$
f_X(x)=\int_{x^2}^{x}6\,dy=6(x-x^2)
$$
因此
$$
f_X(x)=
\begin{cases}
6(x-x^2), & 0<x<1,\\
0, & \text{其他}
\end{cases}
$$

求 $Y$ 的边缘概率密度。由 $x^2<y<x$ 可知 $0<y<1$，且固定 $y$ 时
$$
y<x<\sqrt{y}
$$
所以
$$
f_Y(y)=\int_{-\infty}^{+\infty}f_{X,Y}(x,y)\,dx
=\int_y^{\sqrt{y}}6\,dx
=6(\sqrt{y}-y)
$$
因此
$$
f_Y(y)=
\begin{cases}
6(\sqrt{y}-y), & 0<y<1,\\
0, & \text{其他}
\end{cases}
$$


## 习题5

设随机向量$(X,Y)$的概率密度函数为
$$f(x,y) = \begin{cases} \dfrac{c}{(1+x+y)^n}, & x>0, y>0, \\ 0, & \text{其他}, \end{cases}$$
其中 $n>2$，求常数 $c$ 及 $(X,Y)$ 关于 $X,Y$ 的边缘概率密度函数。

解题思路：利用二重积分求常数 $c$，利用积分求边缘概率密度函数。

解答：

(1) $$\int_0^{\infty} \int_0^{\infty} \frac{c}{(1+x+y)^n} \, dx \, dy = c \int_0^{\infty} \frac{(1+x)^{1-n}}{n-1} \, dx = \frac{c}{(n-1)(n-2)} = 1 \implies c = (n-1)(n-2)$$

(2) $$f_X(x) = \int_0^{\infty} \frac{(n-1)(n-2)}{(1+x+y)^n} \, dy = (n-2)(1+x)^{1-n}, \quad x>0$$
同理，$$f_Y(y) = (n-2)(1+y)^{1-n}, \quad y>0$$

## 习题6
设二维随机向量$(X,Y)$的概率密度函数为
$$f(x,y) = \begin{cases} A(6-x-y), & 0<x<2, 2<y<4, \\ 0, & \text{其他}, \end{cases}$$
求：(1) 常数$A$；(2) $P\{X<1, Y<3\}$；(3) $P\{X+Y<4\}$。

解题思路：利用归一性求 $A$。利用联合概率密度函数积分求区域概率。

解答：

二维连续型随机变量，概率密度满足归一性：
$$
\iint_{\mathbb R^2} f(x,y),dxdy = 1
$$
积分区域 $D:(0<x<2),;2<y<4$。

(1) 求常数 $A$

$$
\begin{aligned}
\int_{0}^{2}\int_{2}^{4} A(6-x-y),dy,dx &= 1 \
A\int_{0}^{2}\left[\int_{2}^{4}(6-x-y),dy\right]dx &=1
\end{aligned}
$$
对内层对 $y$ 积分：
$$
\int_{2}^{4}(6-x-y)dy
=\left[(6-x)y-\frac12 y^2\right]*{y=2}^{y=4}
$$
$$
=\big(4(6-x)-8\big)-\big(2(6-x)-2\big)
=2(6-x)-6 = 6-2x
$$
再对 $x$ 积分：
$$
A\int*{0}^{2}(6-2x),dx = A\left[6x-x^2\right]_0^2 = A(12-4)=8A
$$
由 $8A=1 \implies A=\dfrac18$。


(2) 求 $P{X<1,Y<3}$

积分区域：$0<x<1,;2<y<3$
$$
\begin{aligned}
P{X<1,Y<3}
&=\int_{0}^{1}\int_{2}^{3}\frac18(6-x-y),dy,dx \
&=\frac18\int_{0}^{1}\left[\int_{2}^{3}(6-x-y)dy\right]dx
\end{aligned}
$$
内层积分：
$$
\int_{2}^{3}(6-x-y)dy=\left[(6-x)y-\frac12 y^2\right]*{2}^{3}= \frac72 - x
$$
$$
\begin{aligned}
P&=\frac18\int*{0}^{1}\left(\frac72 - x\right)dx
=\frac18\left[\frac72x-\frac12x^2\right]_0^1
=\frac18\left(\frac72-\frac12\right)
=\frac18\cdot 3
=\boldsymbol{\frac{3}{8}}
\end{aligned}
$$

(3) 求 $P{X+Y<4}$

定义域：$(0<x<2),;2<y<4$，且 $y<4-x$

联立：$y$ 的下限 $2$，上限 $4-x$；

$x$ 范围：由 $4-x>2 \Rightarrow x<2$，所以 $(0<x<2)$

$$
\begin{aligned}
P{X+Y<4}
&=\int_{0}^{2}\int_{2}^{4-x}\frac18(6-x-y),dy,dx
\end{aligned}
$$
先对 $y$ 积分：令 $u=6-x-y$
$$
\int_{2}^{4-x}(6-x-y)dy
=\left[(6-x)y-\frac12 y^2\right]_{y=2}^{y=4-x}
=\frac12(4-x)^2 -2
$$

$$
\begin{aligned}
P&=\frac18\int_{0}^{2}\left[\frac12(4-x)^2 -2\right]dx
=\frac18\int_{0}^{2}\left(\frac12x^2-4x+6\right)dx\
&=\frac18\left[\frac16x^3-2x^2+6x\right]_0^2
=\frac{2}{3}
\end{aligned}
$$

## 习题8

已知随机变量$X_1 \sim \begin{pmatrix} -1 & 0 & 1 \\ \frac{1}{4} & \frac{1}{2} & \frac{1}{4} \end{pmatrix}$，随机变量$X_2 \sim \begin{pmatrix} 0 & 1 \\ \frac{1}{2} & \frac{1}{2} \end{pmatrix}$，且$P\{X_1X_2=0\}=1$。

(1) 求$X_1$和$X_2$的联合概率函数；(2) 判断$X_1$和$X_2$是否相互独立？

解题思路：由 $P\{X_1X_2=0\}=1$ 可知 $P(X_1=-1, X_2=1) = P(X_1=1, X_2=1) = 0$。由此推导联合分布。

解答：

由题意：$P\{X_1X_2=0\}=1$

所以当 $X_2=1$ 时，必须有 $X_1=0$，因此 $P(X_1=-1,X_2=1)=0,\qquad P(X_1=1,X_2=1)=0$

又因为$P(X_2=1)=\frac12$

所以$P(X_1=0,X_2=1)=\frac12$

再由边缘分布可得：
$$
P(X_1=-1,X_2=0)=\frac14,\qquad
P(X_1=1,X_2=0)=\frac14
$$

因此联合概率函数为：
$$
P(X_1=x_1,X_2=x_2)=
\begin{cases}
\dfrac14, & (x_1,x_2)=(-1,0),\\[6pt]
\dfrac12, & (x_1,x_2)=(0,1),\\[6pt]
\dfrac14, & (x_1,x_2)=(1,0),\\[6pt]
0, & \text{其他}.
\end{cases}
$$

写成表格形式：
$$
\begin{array}{c|cc}
X_1 \backslash X_2 & 0 & 1\\
\hline
-1 & \dfrac14 & 0\\[6pt]
0 & 0 & \dfrac12\\[6pt]
1 & \dfrac14 & 0
\end{array}
$$

判断独立性：
$$
P(X_1=0,X_2=1)=\frac12
$$
而
$$
P(X_1=0)P(X_2=1)=\frac12\cdot\frac12=\frac14
$$
因为
$$
\frac12\neq \frac14
$$
所以 $X_1$ 与 $X_2$ 不相互独立。

## 习题9
设二维随机向量$(X,Y)$的概率密度函数为
$$f(x,y) = \begin{cases} 1, & 0<x<1, 0<y<2x, \\ 0, & \text{其他}. \end{cases}$$
(1) 求$(X,Y)$分别关于$X$与$Y$的边缘概率密度函数$f_X(x), f_Y(y)$；(2) 求$Z=X+Y$的概率密度函数。

解题思路：利用积分求边缘概率密度。利用卷积公式求 $Z=X+Y$ 的密度函数。

解答：

(1) $$f_X(x) = \int_0^{2x} 1 \, dy = 2x, \quad 0<x<1$$
$$f_Y(y) = \int_{y/2}^1 1 \, dx = 1 - \frac{y}{2}, \quad 0<y<2$$

(2) 卷积公式 $$f_Z(z) = \int_{-\infty}^{\infty} f(x, z-x) \, dx$$

当 $0<z<1$ 时，$$f_Z(z) = \int_{z/3}^{z} 1 \, dx = \frac{2z}{3}$$

当 $1 \le z < 3$ 时，$$f_Z(z) = \int_{z/3}^{1} 1 \, dx = 1 - \frac{z}{3}$$

其他情况 $f_Z(z) = 0$。

## 习题11
设随机变量$Y$的概率密度函数为
$$f_Y(y) = \begin{cases} 5y^4, & 0<y<1, \\ 0, & \text{其他}, \end{cases}$$
在给定$Y=y$的条件下，随机变量$X$的条件概率密度函数为
$$f_{X|Y}(x|y) = \begin{cases} \frac{3x^2}{y^3}, & 0<x<y<1, \\ 0, & \text{其他}, \end{cases}$$
求概率$P\{X>0.5\}$。

解题思路：$f(x,y) = f_{X|Y}(x|y)f_Y(y)$，然后积分求概率。

解答：
由条件概率密度公式，联合密度为
$$
f_{X,Y}(x,y)=f_Y(y)f_{X|Y}(x|y)
$$

要求
$$
P\{X>0.5\}=\iint_{x>0.5} f_{X,Y}(x,y)\,dx\,dy
$$

因为 $0<x<y<1$，且要求 $x>0.5$，所以必须有 $y>0.5$

于是
$$
P\{X>0.5\}
=\int_{0.5}^{1} f_Y(y)\left(\int_{0.5}^{y} f_{X|Y}(x|y)\,dx\right)dy
$$

先计算内层积分：
$$
\int_{0.5}^{y} \frac{3x^2}{y^3}\,dx
=\frac{3}{y^3}\cdot \frac{x^3}{3}\Bigg|_{0.5}^{y}
=\frac{y^3-0.5^3}{y^3}
=1-\frac{1}{8y^3}
$$

再代入 $f_Y(y)=5y^4$，得
$$
P\{X>0.5\}
=\int_{0.5}^{1} 5y^4\left(1-\frac{1}{8y^3}\right)dy
=\int_{0.5}^{1}\left(5y^4-\frac{5}{8}y\right)dy
$$

计算积分：
$$
\int_{0.5}^{1}\left(5y^4-\frac{5}{8}y\right)dy
=\left[y^5-\frac{5}{16}y^2\right]_{0.5}^{1}
$$

代入上下限：
$$
\left(1-\frac{5}{16}\right)
-\left(\frac{1}{32}-\frac{5}{64}\right)
=\frac{11}{16}-\left(-\frac{3}{64}\right)
=\frac{44}{64}+\frac{3}{64}
=\frac{47}{64}
$$

## 习题12
(1) 设随机向量$(X,Y)$的概率密度函数为
$$f(x,y) = \begin{cases} 4xy, & 0 \le x \le 1, 0 \le y \le 1, \\ 0, & \text{其他}, \end{cases}$$
判断$X$与$Y$是否独立？

(2) 设随机向量$(X,Y)$的概率密度函数为
$$f(x,y) = \begin{cases} 8xy, & 0 \le x \le y, 0 \le y \le 1, \\ 0, & \text{其他}, \end{cases}$$
判断$X$与$Y$是否相互独立？

解题思路：求边缘概率密度函数，检查是否 $f(x,y) = f_X(x)f_Y(y)$。

解答：


(1)

$$f(x,y)=\begin{cases}4xy,&0<x<1,0<y<1\\0,&\text{其他}\end{cases}$$
$$
f_X(x)=\int_0^14xy dy=2x,\;0<x<1,\quad f_Y(y)=2y,\;0<y<1
$$

解题思路：几何概型。在单位正方形中求面积。

解答：

设从 $(0,1)$ 中随机取出的两个数为 $X,Y$，则 $(X,Y)$ 在单位正方形 $0<x<1, 0<y<1$ 上服从均匀分布。

## 习题24

已知二维随机向量$(X,Y)$的概率密度函数为
$$f(x,y) = \begin{cases} \frac{1}{2}\sin(x+y), & 0 \le x \le \frac{\pi}{2}, 0 \le y \le \frac{\pi}{2}, \\ 0, & \text{其他}, \end{cases}$$
求$E(X), E(Y), X$与$Y$的协方差矩阵$\Sigma$。

解题思路：利用二重积分求期望和协方差。

解答：

由联合密度关于 \(x,y\) 对称，所以
$$
E(X)=E(Y),\qquad \operatorname{Var}(X)=\operatorname{Var}(Y).
$$

先求
$$
E(X)=\int_0^{\pi/2}\int_0^{\pi/2} x\cdot \frac12\sin(x+y)\,dx\,dy.
$$
利用
$$
\sin(x+y)=\sin x\cos y+\cos x\sin y,
$$
可得
$$
E(X)=\frac12\left[\left(\int_0^{\pi/2}x\sin x\,dx\right)\left(\int_0^{\pi/2}\cos y\,dy\right)
+\left(\int_0^{\pi/2}x\cos x\,dx\right)\left(\int_0^{\pi/2}\sin y\,dy\right)\right].
$$
而
$$
\int_0^{\pi/2}x\sin x\,dx=1,\qquad
\int_0^{\pi/2}x\cos x\,dx=\frac{\pi}{2}-1,
$$
$$
\int_0^{\pi/2}\cos y\,dy=1,\qquad
\int_0^{\pi/2}\sin y\,dy=1.
$$
所以
$$
E(X)=\frac12\left[1+\frac{\pi}{2}-1\right]=\frac{\pi}{4}.
$$
因此
$$
E(X)=\frac{\pi}{4},\qquad E(Y)=\frac{\pi}{4}.
$$

再求
$$
E(X^2)=\int_0^{\pi/2}\int_0^{\pi/2} x^2\cdot \frac12\sin(x+y)\,dx\,dy.
$$
同样分解可得
$$
E(X^2)=\frac12\left[
\left(\int_0^{\pi/2}x^2\sin x\,dx\right)
+\left(\int_0^{\pi/2}x^2\cos x\,dx\right)
\right].
$$
计算得
$$
\int_0^{\pi/2}x^2\sin x\,dx=\pi-2,
$$
$$
\int_0^{\pi/2}x^2\cos x\,dx=\frac{\pi^2}{4}-2.
$$
所以
$$
E(X^2)=\frac12\left[\pi-2+\frac{\pi^2}{4}-2\right]
=\frac{\pi^2}{8}+\frac{\pi}{2}-2.
$$
于是
$$
\operatorname{Var}(X)=E(X^2)-[E(X)]^2
=\frac{\pi^2}{8}+\frac{\pi}{2}-2-\frac{\pi^2}{16}
=\frac{\pi^2}{16}+\frac{\pi}{2}-2.
$$
由对称性，
$$
\operatorname{Var}(Y)=\operatorname{Var}(X)=\frac{\pi^2}{16}+\frac{\pi}{2}-2.
$$

再求
$$
E(XY)=\int_0^{\pi/2}\int_0^{\pi/2} xy\cdot \frac12\sin(x+y)\,dx\,dy.
$$
分解得
$$
E(XY)=\frac12\left[
\left(\int_0^{\pi/2}x\sin x\,dx\right)
\left(\int_0^{\pi/2}y\cos y\,dy\right)
+
\left(\int_0^{\pi/2}x\cos x\,dx\right)
\left(\int_0^{\pi/2}y\sin y\,dy\right)
\right].
$$
代入
$$
\int_0^{\pi/2}x\sin x\,dx=1,\qquad
\int_0^{\pi/2}x\cos x\,dx=\frac{\pi}{2}-1,
$$
得到
$$
E(XY)=\frac12\left[1\cdot\left(\frac{\pi}{2}-1\right)
+\left(\frac{\pi}{2}-1\right)\cdot 1\right]
=\frac{\pi}{2}-1.
$$
因此协方差为
$$
\operatorname{Cov}(X,Y)=E(XY)-E(X)E(Y)
=\left(\frac{\pi}{2}-1\right)-\frac{\pi}{4}\cdot\frac{\pi}{4}
=\frac{\pi}{2}-1-\frac{\pi^2}{16}.
$$

所以协方差矩阵为
$$
\Sigma=
\begin{pmatrix}
\operatorname{Var}(X) & \operatorname{Cov}(X,Y)\\
\operatorname{Cov}(X,Y) & \operatorname{Var}(Y)
\end{pmatrix}.
$$

即
$$
E(X)=E(Y)=\frac{\pi}{4}
$$
$$
\Sigma=
\begin{pmatrix}
\dfrac{\pi^2}{16}+\dfrac{\pi}{2}-2 & \dfrac{\pi}{2}-1-\dfrac{\pi^2}{16}\\[8pt]
\dfrac{\pi}{2}-1-\dfrac{\pi^2}{16} & \dfrac{\pi^2}{16}+\dfrac{\pi}{2}-2
\end{pmatrix}
$$

## 习题29
已知随机向量$(X,Y) \sim N(2,1;9,4;0.5)$，则$D(X-Y)=(\quad)$。

(A) 7  (B) 8  (C) 13  (D) 5

解题思路：记号 $N(\mu_1, \mu_2; \sigma_1^2, \sigma_2^2; \rho)$。

解答：

由二维正态分布记号
$$
(X,Y)\sim N(2,1;9,4;0.5)
$$
可知：
$$
D(X)=9,\qquad D(Y)=4,\qquad \rho=0.5
$$

所以
$$
\operatorname{Cov}(X,Y)=\rho\sqrt{D(X)}\sqrt{D(Y)}
=0.5\times \sqrt{9}\times \sqrt{4}
=0.5\times 3\times 2=3
$$

于是
$$
D(X-Y)=D(X)+D(Y)-2\operatorname{Cov}(X,Y)
=9+4-2\times 3=7
$$

答案选 A。
