### 习题 3 详细解答

#### 第1题
**解题思路**：利用超几何分布求联合概率。第一次取1个球，第二次取2个球，总样本空间。条件概率公式 $P(A|B) = P(AB)/P(B)$。
**解答**：
总球数8个（5白，3红）。$X_1 \in \{0,1\}$，$X_2 \in \{0,1,2\}$。
(1) 联合概率分布列如下：
*   $P(X_1=0, X_2=0) = \frac{3}{8} \times \frac{\binom{2}{2}}{\binom{7}{2}} = \frac{1}{56}$
*   $P(X_1=0, X_2=1) = \frac{3}{8} \times \frac{\binom{5}{1}\binom{2}{1}}{\binom{7}{2}} = \frac{5}{28}$
*   $P(X_1=0, X_2=2) = \frac{3}{8} \times \frac{\binom{5}{2}}{\binom{7}{2}} = \frac{5}{28}$
*   $P(X_1=1, X_2=0) = \frac{5}{8} \times \frac{\binom{3}{2}}{\binom{7}{2}} = \frac{5}{56}$
*   $P(X_1=1, X_2=1) = \frac{5}{8} \times \frac{\binom{4}{1}\binom{3}{1}}{\binom{7}{2}} = \frac{5}{14}$
*   $P(X_1=1, X_2=2) = \frac{5}{8} \times \frac{\binom{4}{2}}{\binom{7}{2}} = \frac{5}{28}$
(2) $P\{X_1=0, X_2 \neq 0\} = P(0,1) + P(0,2) = \frac{5}{28} + \frac{5}{28} = \frac{5}{14}$
$P\{X_1=X_2\} = P(0,0) + P(1,1) = \frac{1}{56} + \frac{5}{14} = \frac{3}{8}$
$P\{X_1X_2=0\} = P(0,0) + P(1,0) = \frac{1}{56} + \frac{5}{56} = \frac{3}{28}$

#### 第2题
**解题思路**：根据联合概率密度函数的归一性求 $c$。计算对立事件的概率：$P(X < \frac{1}{2} \text{ 或 } Y < \frac{1}{2}) = 1 - P(X \ge \frac{1}{2}, Y \ge \frac{1}{2})$。
**解答**：
由 $\int_{0}^{1} \int_{0}^{2} c \, dy \, dx = 2c = 1 \implies c = \frac{1}{2}$。
$P(X \ge \frac{1}{2}, Y \ge \frac{1}{2}) = \int_{1/2}^{1} \int_{1/2}^{2} \frac{1}{2} \, dy \, dx = \frac{1}{2} \times \frac{1}{2} \times \frac{3}{2} = \frac{3}{8}$。
所以所求概率为 $1 - \frac{3}{8} = \frac{5}{8}$。

#### 第3题
**解题思路**：二维标准正态分布的边缘分布仍是标准正态分布。利用积分求边缘概率密度函数。
**解答**：
$f_X(x) = \int_{-\infty}^{\infty} \frac{1}{2\pi} e^{-\frac{x^2+y^2}{2}} \, dy = \frac{1}{\sqrt{2\pi}} e^{-\frac{x^2}{2}} \int_{-\infty}^{\infty} \frac{1}{\sqrt{2\pi}} e^{-\frac{y^2}{2}} \, dy = \frac{1}{\sqrt{2\pi}} e^{-\frac{x^2}{2}}$。
由对称性，$f_Y(y) = \frac{1}{\sqrt{2\pi}} e^{-\frac{y^2}{2}}$。

#### 第4题
**解题思路**：均匀分布的概率密度函数为区域面积的倒数。需先计算区域 $G$ 的面积。
**解答**：
区域 $G: |x+y| > 1, x^2 < y$。由 $x^2 < y$ 且 $x+y > 1$ 可知，该区域向右上方无限延伸（例如 $x$ 可以无限大，$y > x^2$ 且 $y > 1-x$），因此区域 $G$ 的面积为无穷大。**（注：本题题目可能有误，若区域有界，概率密度函数才存在。若按原题，面积无穷大，无法进行均匀分布归一化。）**

#### 第5题
**解题思路**：利用二重积分求常数 $c$。利用积分求边缘概率密度函数。
**解答**：
(1) $\int_0^{\infty} \int_0^{\infty} \frac{c}{(1+x+y)^n} \, dx \, dy = c \int_0^{\infty} \frac{(1+x)^{1-n}}{n-1} \, dx = \frac{c}{(n-1)(n-2)} = 1 \implies c = (n-1)(n-2)$。
(2) $f_X(x) = \int_0^{\infty} \frac{(n-1)(n-2)}{(1+x+y)^n} \, dy = (n-2)(1+x)^{1-n}, \quad x>0$。
同理，$f_Y(y) = (n-2)(1+y)^{1-n}, \quad y>0$。

#### 第6题
**解题思路**：利用归一性求 $A$。利用联合概率密度函数积分求区域概率。
**解答**：
(1) $\int_0^2 \int_2^4 A(6-x-y) \, dy \, dx = A \int_0^2 (6-2x) \, dx = 8A = 1 \implies A = \frac{1}{8}$。
(2) $P\{X<1, Y<3\} = \int_0^1 \int_2^3 \frac{1}{8}(6-x-y) \, dy \, dx = \frac{1}{8} \int_0^1 (3.5-x) \, dx = \frac{3}{8}$。
(3) $P\{X+Y<4\} = \int_0^2 \int_2^{4-x} \frac{1}{8}(6-x-y) \, dy \, dx = \frac{1}{8} \int_0^2 (\frac{x^2}{2} - 4x + 6) \, dx = \frac{2}{3}$。

#### 第7题
**解题思路**：根据联合分布表求边缘概率、条件概率和判断独立性。
**解答**：
(1) $P(X=1) = \frac{1}{9}, P(X=2) = \frac{2}{3}, P(X=3) = \frac{2}{9}$。
$P(Y=0) = \frac{1}{3}, P(Y=1) = \frac{4}{9}, P(Y=2) = \frac{2}{9}$。
(2) $P(X=1) = \frac{1}{9}$。$P(Y=0|X=1) = 1$，$P(Y=1|X=1) = 0$，$P(Y=2|X=1) = 0$。
(3) 因为 $P(X=1, Y=0) = \frac{1}{9}$，但 $P(X=1)P(Y=0) = \frac{1}{9} \times \frac{1}{3} = \frac{1}{27} \neq \frac{1}{9}$，所以 $X$ 与 $Y$ 不独立。

#### 第8题
**解题思路**：由 $P\{X_1X_2=0\}=1$ 可知 $P(X_1=-1, X_2=1) = P(X_1=1, X_2=1) = 0$。由此推导联合分布。
**解答**：
(1) 联合分布律为：
$P(X_1=-1, X_2=0) = \frac{1}{4}$，$P(X_1=-1, X_2=1) = 0$
$P(X_1=0, X_2=0) = 0$，$P(X_1=0, X_2=1) = \frac{1}{2}$
$P(X_1=1, X_2=0) = \frac{1}{4}$，$P(X_1=1, X_2=1) = 0$
(2) 因为 $P(X_1=-1, X_2=0) = \frac{1}{4} \neq P(X_1=-1)P(X_2=0) = \frac{1}{8}$，故不独立。

#### 第9题
**解题思路**：利用积分求边缘概率密度。利用卷积公式求 $Z=X+Y$ 的密度函数。
**解答**：
(1) $f_X(x) = \int_0^{2x} 1 \, dy = 2x, \quad 0<x<1$；$f_Y(y) = \int_{y/2}^1 1 \, dx = 1 - \frac{y}{2}, \quad 0<y<2$。
(2) 卷积公式 $f_Z(z) = \int_{-\infty}^{\infty} f(x, z-x) \, dx$。
当 $0<z<1$ 时，$f_Z(z) = \int_{z/3}^{z} 1 \, dx = \frac{2z}{3}$。
当 $1 \le z < 3$ 时，$f_Z(z) = \int_{z/3}^{1} 1 \, dx = 1 - \frac{z}{3}$。
其他情况 $f_Z(z) = 0$。

#### 第10题
**解题思路**：利用条件概率密度公式 $f_{Y|X}(y|x) = \frac{f(x,y)}{f_X(x)}$。
**解答**：
$f_X(x) = \int_0^1 (x+y) \, dy = x + \frac{1}{2}, \quad 0<x<1$。
条件概率密度：$f_{Y|X}(y|x) = \frac{x+y}{x+1/2}, \quad 0<y<1$。
条件分布函数：$F_{Y|X}(y|x) = \int_0^y \frac{x+t}{x+1/2} \, dt = \frac{xy + y^2/2}{x+1/2}, \quad 0<y<1$。

#### 第11题
**解题思路**：$f(x,y) = f_{X|Y}(x|y)f_Y(y)$，然后积分求概率。
**解答**：
$f(x,y) = \frac{3x^2}{y^3} \cdot 5y^4 = 15x^2y, \quad 0<x<y<1$。
$P(X>0.5) = \int_{0.5}^1 \int_{0.5}^y 15x^2y \, dx \, dy = \int_{0.5}^1 (5y^4 - \frac{5}{8}y) \, dy = \frac{47}{64}$。

#### 第12题
**解题思路**：求边缘概率密度函数，检查是否 $f(x,y) = f_X(x)f_Y(y)$。
**解答**：
(1) $f_X(x) = 2x, f_Y(y) = 2y$。$f_X(x)f_Y(y) = 4xy = f(x,y)$，且区域为矩形，故独立。
(2) $f_X(x) = 4x(1-x^2), f_Y(y) = 4y^3$。$f_X(x)f_Y(y) \neq f(x,y)$，故不独立。

#### 第13题
**解题思路**：独立随机变量联合密度为边缘密度之积。方程有实根的条件是判别式 $\Delta \ge 0$。
**解答**：
(1) $f(x,y) = \frac{1}{2}e^{-y/2}, \quad 0<x<1, y>0$。
(2) $\Delta = (2X)^2 - 4Y \ge 0 \implies X^2 \ge Y$。
$P(X^2 \ge Y) = \int_0^1 \int_0^{x^2} \frac{1}{2}e^{-y/2} \, dy \, dx = \int_0^1 (1 - e^{-x^2/2}) \, dx = 1 - \int_0^1 e^{-x^2/2} \, dx$。（注：该积分无初等原函数，可用标准正态分布函数表示）。

#### 第14题
**解题思路**：独立随机变量和的密度用卷积公式，商的密度用变换公式 $f_{X/Y}(z) = \int |y|f_X(yz)f_Y(y) \, dy$。
**解答**：
$f_X(x) = e^{-x}, x>0$；$f_Y(y) = e^{-y}, y>0$。
$X+Y$ 的密度：$f_{X+Y}(z) = \int_0^z e^{-x}e^{-(z-x)} \, dx = ze^{-z}, \quad z>0$。
